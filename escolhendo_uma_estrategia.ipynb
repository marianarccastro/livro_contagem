# Escolhendo uma estratégia

Até aqui, você conheceu diferentes maneiras de organizar e resolver problemas de contagem. Neste capítulo, essas estratégias aparecerão misturadas.

Antes de calcular, será preciso analisar cada situação e decidir como organizar a contagem. Em alguns problemas, uma única estratégia será suficiente. Em outros, será necessário combinar ideias estudadas ao longo do livro.

O desafio agora não é reconhecer o “tipo” de problema, mas decidir **por onde começar**.

## Antes de começar

Ao analisar um problema, algumas perguntas podem ajudar:

- Quais decisões precisam ser tomadas?
- O que muda depois de cada escolha?
- As decisões representam alternativas ou etapas sucessivas?
- A ordem altera o resultado?
- Algum resultado está sendo contado mais de uma vez?
- Existe alguma restrição? Em qual decisão ela interfere?
- Há uma decisão mais restrita pela qual seja conveniente começar?
- É útil dividir o problema em casos?
- É mais simples contar diretamente ou contar o contrário?
- É necessário combinar mais de uma estratégia?

Essas perguntas não formam um roteiro que precisa ser seguido em todos os problemas. Elas servem para ajudar você a observar a estrutura de cada situação e escolher uma forma de organizar a contagem.
