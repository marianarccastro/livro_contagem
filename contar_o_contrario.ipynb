Podemos então separar todas as senhas em dois grupos:

$$
\text{Todas as senhas}
=
\text{senhas com pelo menos um 7}
+
\text{senhas sem nenhum 7}
$$

Sem nenhuma restrição, temos 10 escolhas para cada uma das quatro posições:

$$
10 \cdot 10 \cdot 10 \cdot 10 = 10^4
$$

Para formar uma senha **sem nenhum algarismo 7**, temos apenas 9 escolhas em cada posição:

$$
9 \cdot 9 \cdot 9 \cdot 9 = 9^4
$$

Portanto,

$$
10^4 - 9^4 = 10000 - 6561 = 3439
$$

Assim, existem **3439 senhas com pelo menos um algarismo 7**.

:::{important}
Quando um problema pede **pelo menos um** elemento com determinada característica, pode ser útil investigar o caso contrário: **nenhum** elemento com essa característica.
:::

## Escolhendo sabores

Uma sorveteria oferece 8 sabores, sendo 3 sabores de frutas.

Uma pessoa deseja escolher 3 sabores diferentes.

**De quantas maneiras essa escolha pode ser feita se pelo menos um dos sabores escolhidos deve ser de fruta?**

Antes de calcular, pense:

> É mais simples contar diretamente as escolhas com pelo menos um sabor de fruta ou contar primeiro as escolhas sem nenhum sabor de fruta?

```python
%%html

<textarea
  style="
    width: 100%;
    height: 100px;
    border: 1px solid #aaa;
    border-radius: 8px;
    padding: 10px;
    font-family: system-ui, sans-serif;
    font-size: 15px;
    box-sizing: border-box;
  "
  placeholder="Registre aqui a estratégia que você usaria..."
></textarea>
```

Podemos começar contando **todas as escolhas possíveis de 3 sabores**.

Se considerarmos a ordem das escolhas, temos:

$$
8 \cdot 7 \cdot 6
$$

Mas escolher, por exemplo, chocolate, morango e baunilha representa o mesmo grupo que escolher morango, baunilha e chocolate.

Como a ordem dos 3 sabores não altera a escolha, cada grupo foi contado $3!$ vezes.

Assim, o número total de escolhas é:

$$
\frac{8 \cdot 7 \cdot 6}{3!}=56
$$

Agora vamos contar as escolhas que **não possuem nenhum sabor de fruta**.

Como 3 dos 8 sabores são de frutas, restam 5 sabores que não são de frutas.

Escolhendo 3 deles:

$$
\frac{5 \cdot 4 \cdot 3}{3!}=10
$$

Portanto, das 56 escolhas possíveis, 10 não possuem nenhum sabor de fruta.

Logo:

$$
56-10=46
$$

Assim, existem **46 maneiras de escolher 3 sabores com pelo menos um sabor de fruta**.

:::{important}
Neste problema, contar diretamente as escolhas com **pelo menos um sabor de fruta** exigiria dividir a contagem em vários casos.

Contando o contrário, precisamos considerar apenas as escolhas com **nenhum sabor de fruta**.
:::

A estratégia continua sendo a mesma:

:::{dropdown} Precisa de uma pista?

Em vez de contar diretamente os números que possuem pelo menos um $5$, conte primeiro os números de três algarismos que **não possuem nenhum $5$**.

Lembre-se de que o primeiro algarismo não pode ser $0$.

:::

Existem $9$ possibilidades para o algarismo das centenas, $10$ para o das dezenas e $10$ para o das unidades.

Assim, existem:

$$
9 \cdot 10 \cdot 10 = 900
$$

números de três algarismos.

Agora vamos contar os números que **não possuem nenhum algarismo $5$**.

Para a centena, não podemos usar $0$ nem $5$, então temos $8$ possibilidades.

Para a dezena e para a unidade, podemos usar qualquer algarismo, exceto $5$, totalizando $9$ possibilidades em cada posição.

Logo:

$$
8 \cdot 9 \cdot 9 = 648
$$

números não possuem nenhum algarismo $5$.

Portanto:

$$
900 - 648 = 252
$$

Assim, existem **252 números de três algarismos com pelo menos um algarismo $5$**.

:::{important}
Contar o contrário pode simplificar o problema, mas ainda precisamos considerar com atenção as restrições de cada escolha.

Neste caso, o primeiro algarismo não pode ser $0$.
:::
