## Elementos que precisam ficar juntos

Na situação anterior, a restrição determinava **onde** Ana poderia ficar.

Agora, vamos analisar um tipo diferente de condição.

Cinco estudantes — Ana, Bruno, Caio, Dani e Eva — vão se posicionar em fila para tirar uma foto.

Sem nenhuma restrição, poderíamos organizar os cinco estudantes de $5!=120$ maneiras diferentes.

Mas Ana e Bruno querem ficar **juntos**.

De quantas maneiras diferentes os cinco estudantes podem se organizar respeitando essa condição?

Uma forma de começar é pensar em Ana e Bruno como se, temporariamente, formassem um único bloco:

**[Ana Bruno]  Caio  Dani  Eva**

Assim, em vez de organizar cinco elementos separadamente, passamos a organizar $4$ elementos.

Antes de continuar, pense:

* De quantas maneiras esses $4$ elementos podem ser organizados?
* Ao tratar Ana e Bruno como um único bloco, já estamos considerando todas as possibilidades?
* O que ainda pode mudar dentro desse bloco?

Acompanhe a animação e observe como a condição de manter Ana e Bruno juntos modifica a organização da fila.

:::{iframe} https://www.youtube.com/embed/kqfnqJZslO0?si=GOwaWWfMC2ReUDpd
:width: 100%
:title: Elementos juntos em uma fila
:::

*Ana e Bruno podem ser tratados temporariamente como um único bloco.*

Depois de tratar Ana e Bruno como um único bloco, passamos a organizar $4$ elementos:

**[Ana e Bruno]**, Caio, Dani e Eva.

Esses $4$ elementos podem ser organizados de

$4!=24$

maneiras.

Mas, dentro do bloco, Ana e Bruno também podem trocar de posição:

**[Ana | Bruno]** ou **[Bruno | Ana]**.

Portanto, para cada uma das $24$ organizações dos quatro elementos, existem $2$ maneiras de organizar Ana e Bruno.

Assim,

$4!\cdot 2!=24\cdot 2=48$.

Logo, existem **$48$ maneiras** de organizar os cinco estudantes mantendo Ana e Bruno juntos.

> Quando alguns elementos precisam ficar juntos, podemos tratá-los temporariamente como um único bloco e, depois, considerar as possíveis organizações dentro desse bloco.

## Para finalizar

Quando um problema apresenta uma restrição, algumas possibilidades deixam de ser válidas.

Por isso, antes de contar, é importante identificar **qual escolha é afetada pela condição**.

Em muitos casos, começar por essa escolha torna a contagem mais simples e evita considerar possibilidades que depois precisariam ser descartadas.

> **Em problemas com restrições, comece analisando a escolha mais limitada.**
