```python
from IPython.display import HTML, display

display(HTML("""
<div id="interacao-restricao-fila" style="
    font-family: system-ui, sans-serif;
    max-width: 900px;
    margin: 20px auto;
    padding: 20px;
    border: 1px solid #d6d9dc;
    border-radius: 12px;
    background-color: #f8f9fa;
">

    <style>
        #interacao-restricao-fila .titulo-etapa {
            font-size: 1.08rem;
            font-weight: 700;
            margin-bottom: 14px;
        }

        #interacao-restricao-fila .fila {
            display: flex;
            gap: 10px;
            flex-wrap: wrap;
            margin: 12px 0 16px 0;
        }

        #interacao-restricao-fila .posicao {
            width: 92px;
            min-height: 54px;
            border: 2px solid #9ca3af;
            border-radius: 9px;
            background: white;
            display: flex;
            align-items: center;
            justify-content: center;
            font-weight: 600;
            cursor: pointer;
            transition: 0.2s;
            box-sizing: border-box;
        }

        #interacao-restricao-fila .posicao:hover {
            background: #f1f3f5;
            transform: translateY(-1px);
        }

        #interacao-restricao-fila .posicao.ana {
            background: #d9d2f3;
            border-color: #7e57c2;
            color: #3f2d68;
            cursor: default;
        }

        #interacao-restricao-fila .posicao.bloqueada {
            cursor: default;
        }

        #interacao-restricao-fila .mensagem {
            margin: 12px 0;
            padding: 10px 12px;
            border-radius: 8px;
            display: none;
        }

        #interacao-restricao-fila .mensagem.erro {
            display: block;
            background: #fdecec;
            border: 1px solid #e5a4a4;
        }

        #interacao-restricao-fila .mensagem.acerto {
            display: block;
            background: #e8f5e9;
            border: 1px solid #9fc7a3;
        }

        #interacao-restricao-fila .etapa {
            margin-top: 20px;
        }

        #interacao-restricao-fila input {
            width: 90px;
            padding: 9px;
            border: 1px solid #aeb4ba;
            border-radius: 8px;
            font-size: 1rem;
            text-align: center;
        }

        #interacao-restricao-fila button {
            padding: 9px 16px;
            margin-left: 8px;
            border: none;
            border-radius: 8px;
            background: #5c6bc0;
            color: white;
            cursor: pointer;
            font-size: 0.95rem;
        }

        #interacao-restricao-fila button:hover {
            opacity: 0.9;
        }

        #interacao-restricao-fila .painel-filas {
            display: grid;
            grid-template-columns: repeat(2, minmax(280px, 1fr));
            gap: 24px;
            margin-top: 20px;
        }

        #interacao-restricao-fila .grupo-filas {
            background: white;
            border: 1px solid #d6d9dc;
            border-radius: 10px;
            padding: 14px;
        }

        #interacao-restricao-fila .grupo-filas h4 {
            margin-top: 0;
        }

        #interacao-restricao-fila .fila-final {
            display: flex;
            gap: 5px;
            margin: 8px 0;
            flex-wrap: nowrap;
        }

        #interacao-restricao-fila .pessoa {
            padding: 6px 8px;
            border-radius: 6px;
            background: #e9ecef;
            font-size: 0.88rem;
            min-width: 48px;
            text-align: center;
        }

        #interacao-restricao-fila .pessoa.ana-final {
            background: #d9d2f3;
            color: #3f2d68;
            font-weight: 700;
        }

        #interacao-restricao-fila .conclusao {
            margin-top: 22px;
            padding: 14px;
            background: #eef0fb;
            border-left: 4px solid #5c6bc0;
            border-radius: 7px;
        }

        @media (max-width: 700px) {
            #interacao-restricao-fila .painel-filas {
                grid-template-columns: 1fr;
            }

            #interacao-restricao-fila .posicao {
                width: 70px;
            }
        }
    </style>

    <!-- ETAPA 1 -->
    <div id="etapa1">
        <div class="titulo-etapa">Escolha uma posição para Ana.</div>

        <div class="fila" id="fila-escolha-1">
            <div class="posicao" data-pos="1">1</div>
            <div class="posicao" data-pos="2">2</div>
            <div class="posicao" data-pos="3">3</div>
            <div class="posicao" data-pos="4">4</div>
        </div>

        <div id="mensagem1" class="mensagem"></div>

        <div id="segunda-escolha" style="display:none;">
            <p><strong>Existe mais alguma posição em que podemos colocá-la?</strong></p>

            <div class="fila" id="fila-escolha-2">
                <div class="posicao" data-pos="1">1</div>
                <div class="posicao" data-pos="2">2</div>
                <div class="posicao" data-pos="3">3</div>
                <div class="posicao" data-pos="4">4</div>
            </div>

            <div id="mensagem2" class="mensagem"></div>
        </div>
    </div>

    <!-- ETAPA 2 -->
    <div id="etapa2" class="etapa" style="display:none;">
        <div class="titulo-etapa">
            De quantas maneiras podemos colocar as outras 3 pessoas na fila?
        </div>

        <p>
            Depois de escolher a posição de Ana, ainda precisamos organizar
            Bruno, Caio e Dani nas três posições restantes.
        </p>

        <input
            type="number"
            id="resposta-organizacao"
            min="0"
            placeholder="?"
        >

        <button id="verificar-resposta">Verificar</button>

        <div id="mensagem-resposta" class="mensagem"></div>
    </div>

    <!-- ETAPA 3 -->
    <div id="etapa3" class="etapa" style="display:none;">
        <div class="titulo-etapa">Veja todas as possibilidades.</div>

        <div class="painel-filas">

            <div class="grupo-filas">
                <h4>Ana na primeira posição — 6 maneiras</h4>
                <div id="grupo-esquerda"></div>
            </div>

            <div class="grupo-filas">
                <h4>Ana na última posição — 6 maneiras</h4>
                <div id="grupo-direita"></div>
            </div>

        </div>

        <div class="conclusao">
            Para escolher a posição de Ana, temos <strong>2 possibilidades</strong>.
            Para cada uma delas, podemos organizar as outras três pessoas de
            <strong>3! = 6 maneiras</strong>.
            <br><br>
            Assim, <strong>2 · 3! = 2 · 6 = 12</strong>.
            <br><br>
            Portanto, existem <strong>12 maneiras</strong> de formar a fila
            respeitando a condição.
        </div>
    </div>

</div>

<script>
(function() {

    const raiz = document.getElementById("interacao-restricao-fila");

    const fila1 = raiz.querySelector("#fila-escolha-1");
    const fila2 = raiz.querySelector("#fila-escolha-2");

    const mensagem1 = raiz.querySelector("#mensagem1");
    const mensagem2 = raiz.querySelector("#mensagem2");

    const segundaEscolha = raiz.querySelector("#segunda-escolha");
    const etapa2 = raiz.querySelector("#etapa2");
    const etapa3 = raiz.querySelector("#etapa3");

    let primeiraPosicao = null;

    function mostrarMensagem(elemento, texto, tipo) {
        elemento.className = "mensagem " + tipo;
        elemento.innerHTML = texto;
    }

    function limparMensagem(elemento) {
        elemento.className = "mensagem";
        elemento.innerHTML = "";
    }

    // -------------------------------
    // PRIMEIRA ESCOLHA
    // -------------------------------

    fila1.querySelectorAll(".posicao").forEach(function(caixa) {

        caixa.addEventListener("click", function() {

            if (primeiraPosicao !== null) return;

            const pos = Number(caixa.dataset.pos);

            if (pos !== 1 && pos !== 4) {

                mostrarMensagem(
                    mensagem1,
                    "Essa posição não atende à condição: Ana precisa ficar em uma das pontas.",
                    "erro"
                );

                return;
            }

            limparMensagem(mensagem1);

            primeiraPosicao = pos;

            caixa.textContent = "Ana";
            caixa.classList.add("ana");

            fila1.querySelectorAll(".posicao").forEach(function(item) {
                item.classList.add("bloqueada");
            });

            segundaEscolha.style.display = "block";

        });

    });

    // -------------------------------
    // SEGUNDA ESCOLHA
    // -------------------------------

    fila2.querySelectorAll(".posicao").forEach(function(caixa) {

        caixa.addEventListener("click", function() {

            const pos = Number(caixa.dataset.pos);

            if (pos !== 1 && pos !== 4) {

                mostrarMensagem(
                    mensagem2,
                    "Essa posição também não atende à condição: Ana precisa ficar em uma das pontas.",
                    "erro"
                );

                return;
            }

            if (pos === primeiraPosicao) {

                mostrarMensagem(
                    mensagem2,
                    "Essa posição já foi encontrada. Procure outra possibilidade.",
                    "erro"
                );

                return;
            }

            limparMensagem(mensagem2);

            caixa.textContent = "Ana";
            caixa.classList.add("ana");

            fila2.querySelectorAll(".posicao").forEach(function(item) {
                item.classList.add("bloqueada");
            });

            mostrarMensagem(
                mensagem2,
                "Existem <strong>2 maneiras</strong> de escolher a posição de Ana.",
                "acerto"
            );

            etapa2.style.display = "block";

        });

    });

    // -------------------------------
    // RESPOSTA: 3! = 6
    // -------------------------------

    const campoResposta = raiz.querySelector("#resposta-organizacao");
    const botaoVerificar = raiz.querySelector("#verificar-resposta");
    const mensagemResposta = raiz.querySelector("#mensagem-resposta");

    function verificarResposta() {

        const resposta = Number(campoResposta.value);

        if (resposta === 6) {

            mostrarMensagem(
                mensagemResposta,
                "Isso mesmo! Para cada posição escolhida para Ana, as outras três pessoas podem ser organizadas de <strong>3! = 6 maneiras</strong>.",
                "acerto"
            );

            etapa3.style.display = "block";

            campoResposta.disabled = true;
            botaoVerificar.disabled = true;

            criarFilas();

        } else {

            mostrarMensagem(
                mensagemResposta,
                "Pense nas escolhas sucessivas: quantas opções temos para a primeira posição livre? E para a posição seguinte?",
                "erro"
            );

        }

    }

    botaoVerificar.addEventListener("click", verificarResposta);

    campoResposta.addEventListener("keydown", function(evento) {
        if (evento.key === "Enter") {
            verificarResposta();
        }
    });

    // -------------------------------
    // CRIAR AS 12 FILAS
    // -------------------------------

    function gerarPermutacoes(lista) {

        if (lista.length === 1) {
            return [lista];
        }

        let resultado = [];

        lista.forEach(function(item, indice) {

            const restantes = lista
                .slice(0, indice)
                .concat(lista.slice(indice + 1));

            const permutacoesRestantes = gerarPermutacoes(restantes);

            permutacoesRestantes.forEach(function(permutacao) {
                resultado.push([item].concat(permutacao));
            });

        });

        return resultado;
    }

    function montarFila(pessoas) {

        const divFila = document.createElement("div");
        divFila.className = "fila-final";

        pessoas.forEach(function(nome) {

            const pessoa = document.createElement("div");
            pessoa.className = "pessoa";

            if (nome === "Ana") {
                pessoa.classList.add("ana-final");
            }

            pessoa.textContent = nome;

            divFila.appendChild(pessoa);

        });

        return divFila;
    }

    let filasCriadas = false;

    function criarFilas() {

        if (filasCriadas) return;
        filasCriadas = true;

        const grupoEsquerda = raiz.querySelector("#grupo-esquerda");
        const grupoDireita = raiz.querySelector("#grupo-direita");

        const outrasPessoas = ["Bruno", "Caio", "Dani"];

        const permutacoes = gerarPermutacoes(outrasPessoas);

        permutacoes.forEach(function(permutacao) {

            const filaAnaInicio = ["Ana"].concat(permutacao);
            const filaAnaFim = permutacao.concat(["Ana"]);

            grupoEsquerda.appendChild(
                montarFila(filaAnaInicio)
            );

            grupoDireita.appendChild(
                montarFila(filaAnaFim)
            );

        });

    }

})();
</script>
"""))
```
