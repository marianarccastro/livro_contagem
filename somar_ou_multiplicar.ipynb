O raciocínio utilizado nesses exemplos é chamado de **Princípio Aditivo da Contagem**.


```{admonition} Princípio Aditivo da Contagem
:class: important

Se uma escolha pode ser feita:

- de $x_1$ maneiras em um primeiro caso;
- de $x_2$ maneiras em um segundo caso;
- $\ldots$
- de $x_n$ maneiras em um $n$-ésimo caso;

sem que uma mesma possibilidade pertença a mais de um caso, então o número total de maneiras de fazer a escolha é

$$x_1+x_2+\cdots+x_n.$$
```


A condição de que os casos não tenham possibilidades em comum é importante. Se a mesma possibilidade aparecer em dois grupos, ela será contada duas vezes.

Neste capítulo, sempre que dividirmos uma contagem em casos, organizaremos os casos de modo que cada possibilidade seja incluída uma única vez.

## Quando soma e multiplicação aparecem juntas

Em muitos problemas, não basta escolher entre somar ou multiplicar. Podemos precisar das duas operações.

Uma sorveteria oferece dois tipos de pedido:

- **casquinha:** escolha 1 entre 6 sabores;
- **copo:** escolha 1 entre 6 sabores e 1 entre 3 coberturas.

Uma pessoa fará apenas um pedido: uma casquinha **ou** um copo.

Antes de continuar, responda:

**a)** Quantas casquinhas diferentes podem ser pedidas?

**b)** Quantos copos diferentes podem ser montados?

**c)** Como podemos reunir os dois tipos de pedido?


````python
%%html
<p><b>Registre seu raciocínio:</b></p>
<textarea style="width: 100%; height: 100px; border: 1px solid #ccc; border-radius: 4px;"></textarea>
````

```{admonition} Compare sua estratégia
:class: dropdown

Para pedir uma casquinha, há apenas uma escolha: o sabor. Portanto, existem

$$6$$

casquinhas diferentes.

Para montar um copo, escolhemos um sabor **e depois** uma cobertura:

$$6\cdot3=18.$$

Por fim, o pedido será uma casquinha **ou** um copo. Os dois casos são separados, então somamos:

$$6+6\cdot3=6+18=24.$$

Há 24 pedidos diferentes.
```


Observe a organização do raciocínio:

1. dividimos o problema em casos;
2. contamos cada caso usando a estratégia adequada;
3. somamos os resultados, pois um pedido não pode ser simultaneamente uma casquinha e um copo.

Essa forma de organizar a contagem é chamada de **divisão em casos**.


## Dividindo uma contagem em casos

Uma escola criará códigos para identificar materiais. Cada código deverá seguir um destes formatos:

- **Formato 1:** uma letra entre A, B, C e D, seguida de um algarismo de 0 a 9;
- **Formato 2:** duas letras, escolhidas entre A, B, C e D, com repetição permitida.

Quantos códigos diferentes podem ser criados?


No primeiro formato, há duas etapas:

$$4\cdot10=40.$$

No segundo formato, também há duas etapas. Como a repetição é permitida, há 4 opções para cada posição:

$$4\cdot4=16.$$

Como cada código pertence a apenas um dos formatos, somamos os resultados:

$$40+16=56.$$

Portanto, podem ser criados 56 códigos diferentes.


```{admonition} Divisão em casos
:class: important

Quando as possibilidades podem ser separadas em casos que não se sobrepõem:

1. identificamos os casos;
2. contamos as possibilidades de cada caso;
3. somamos os resultados.

Dentro de cada caso, pode ser necessário usar o Princípio Fundamental da Contagem.
```


## Um cuidado ao dividir em casos

Uma pessoa escolherá um número inteiro de 1 a 20 que seja múltiplo de 2 **ou** múltiplo de 5.

Poderíamos contar:

- 10 múltiplos de 2;
- 4 múltiplos de 5.

Mas a soma $10+4$ não dá a resposta correta. Por quê?

```{admonition} Observe a sobreposição
:class: dropdown

Os números 10 e 20 são múltiplos de 2 e também de 5. Portanto, seriam contados duas vezes em $10+4$.

Para usar diretamente o princípio aditivo, podemos reorganizar os casos:

- múltiplos de 2: 10 números;
- múltiplos de 5 que não são múltiplos de 2: 5 e 15.

Assim, os casos não se sobrepõem e temos

$$10+2=12.$$

Neste livro, retomaremos situações de sobreposição mais adiante. Por enquanto, ao dividir um problema em casos, verifique se cada possibilidade foi contada uma única vez.
```


## Estratégia de leitura

Ao encontrar um novo problema de contagem, as perguntas abaixo podem ajudar:

1. **Como uma possibilidade completa é formada?**
2. **Há decisões sucessivas?** Nesse caso, a multiplicação pode ser adequada.
3. **Há alternativas ou formatos diferentes?** Nesse caso, pode ser útil dividir em casos e somar.
4. **Os casos se sobrepõem?** Se sim, é preciso reorganizá-los para evitar contagens repetidas.

Não se trata de escolher uma operação por uma palavra isolada, mas de compreender a estrutura do problema.
