````python
%%html

<div id="interacao-ordem-cap4" style="
    font-family: system-ui, sans-serif;
    max-width: 900px;
    margin: 20px auto;
    color: #222;
">

    <!-- ========================================================= -->
    <!-- ÁRVORE 1 -->
    <!-- ========================================================= -->

    <div style="
        border: 1px solid #d6d9dc;
        border-radius: 10px;
        padding: 18px;
        margin-bottom: 28px;
        background: #fafafa;
    ">
        <h3 style="margin-top:0;">Situação 1 — Presidente e vice-presidente</h3>

        <p>
            Quatro estudantes — <b>Ana, Bruno, Caio e Davi</b> — participam de uma eleição.
            Um deles será escolhido presidente e outro vice-presidente.
        </p>

        <p style="margin-bottom:8px;">
            <b>Passe o mouse sobre um resultado da árvore.</b>
        </p>

        <div style="overflow-x:auto;">
            <svg
                id="arvore-cargos-cap4"
                viewBox="0 0 740 510"
                style="width:100%;"
                aria-label="Diagrama de árvore para escolha de presidente e vice-presidente"
            ></svg>
        </div>

        <div id="resultado-cargos-cap4" style="
            min-height: 54px;
            margin-top: 12px;
            padding: 12px 14px;
            border-radius: 8px;
            background: white;
            border: 1px solid #ddd;
        ">
            Passe o mouse sobre um dos resultados para interpretar o caminho.
        </div>
    </div>


    <!-- ========================================================= -->
    <!-- ÁRVORE 2 -->
    <!-- ========================================================= -->

    <div style="
        border: 1px solid #d6d9dc;
        border-radius: 10px;
        padding: 18px;
        background: #fafafa;
    ">
        <h3 style="margin-top:0;">Situação 2 — Formação de uma dupla</h3>

        <p>
            Agora queremos escolher <b>duas das mesmas quatro pessoas</b>
            para formar uma dupla.
        </p>

        <div id="instrucao-duplas-cap4" style="
            padding: 12px 14px;
            margin: 14px 0;
            border-radius: 8px;
            background: #eef5ff;
            border-left: 4px solid #4f7db8;
        ">
            <b>1.</b> Clique em um resultado da árvore.
        </div>

        <div style="overflow-x:auto;">
            <svg
                id="arvore-duplas-cap4"
                viewBox="0 0 740 510"
                style="width:100%;"
                aria-label="Diagrama de árvore para formação de duplas"
            ></svg>
        </div>

        <div id="feedback-duplas-cap4" style="
            min-height: 54px;
            margin-top: 12px;
            padding: 12px 14px;
            border-radius: 8px;
            background: white;
            border: 1px solid #ddd;
        ">
            Escolha um resultado para começar.
        </div>

        <div id="resumo-duplas-cap4" style="
            display:none;
            margin-top:16px;
            padding:14px;
            border-radius:8px;
            border:1px solid #ccc;
            background:white;
        "></div>

        <button id="reiniciar-duplas-cap4" style="
            margin-top: 16px;
            padding: 9px 16px;
            border: none;
            border-radius: 6px;
            background: #555;
            color: white;
            cursor: pointer;
            font-size: 14px;
        ">
            Reiniciar
        </button>
    </div>

</div>


<style>

#interacao-ordem-cap4 .ramo {
    stroke: #adb5bd;
    stroke-width: 2;
    fill: none;
    transition: all 0.2s ease;
}

#interacao-ordem-cap4 .ramo-destaque {
    stroke: #333;
    stroke-width: 4;
}

#interacao-ordem-cap4 .no-pessoa {
    fill: white;
    stroke: #777;
    stroke-width: 1.5;
}

#interacao-ordem-cap4 .texto-pessoa {
    font-size: 15px;
    dominant-baseline: middle;
    text-anchor: middle;
    pointer-events: none;
}

#interacao-ordem-cap4 .resultado {
    fill: white;
    stroke: #888;
    stroke-width: 1.5;
    cursor: pointer;
    transition: all 0.2s ease;
}

#interacao-ordem-cap4 .resultado:hover {
    stroke-width: 3;
}

#interacao-ordem-cap4 .texto-resultado {
    font-size: 14px;
    dominant-baseline: middle;
    text-anchor: middle;
    pointer-events: none;
}

#interacao-ordem-cap4 .resultado-selecionado {
    stroke: #222;
    stroke-width: 4;
}

</style>


<script>
(function () {

    const pessoas = [
        { nome: "Ana", inicial: "A" },
        { nome: "Bruno", inicial: "B" },
        { nome: "Caio", inicial: "C" },
        { nome: "Davi", inicial: "D" }
    ];

    const NS = "http://www.w3.org/2000/svg";

    const coresDuplas = {
        "AB": "#ffd6a5",
        "AC": "#caffbf",
        "AD": "#bde0fe",
        "BC": "#ffc8dd",
        "BD": "#d8c4f1",
        "CD": "#fff3b0"
    };

    function criarSVGElemento(tipo, atributos = {}) {
        const el = document.createElementNS(NS, tipo);

        Object.entries(atributos).forEach(([chave, valor]) => {
            el.setAttribute(chave, valor);
        });

        return el;
    }


    function chaveDupla(a, b) {
        return [a, b].sort().join("");
    }


    function gerarResultados() {

        const resultados = [];

        pessoas.forEach(primeira => {

            pessoas.forEach(segunda => {

                if (primeira.inicial !== segunda.inicial) {

                    resultados.push({
                        primeira,
                        segunda,
                        codigo: primeira.inicial + segunda.inicial,
                        dupla: chaveDupla(
                            primeira.inicial,
                            segunda.inicial
                        )
                    });

                }

            });

        });

        return resultados;
    }


    const resultados = gerarResultados();


    /* ========================================================= */
    /* FUNÇÃO GERAL PARA DESENHAR ÁRVORE                         */
    /* ========================================================= */

    function desenharArvore(svg, tipo) {
    
        svg.innerHTML = "";
    
        const inicioX = 45;
        const primeiraEscolhaX = 220;
        const segundaEscolhaX = 500;
        const resultadoFinalX = 590;
    
        const topo = 35;
        const espacamento = 38;
    
        /*
        Cada pessoa ocupa um bloco com três resultados.
        O nó da primeira escolha fica no centro desses três.
        */
    
        pessoas.forEach((pessoa, indicePessoa) => {
    
            const resultadosPessoa = resultados.filter(
                r => r.primeira.inicial === pessoa.inicial
            );
    
            const primeiroIndice = indicePessoa * 3;
    
            const yCentral =
                topo +
                (primeiroIndice + 1) * espacamento;
    
    
            /*
            Ramo do início até a primeira escolha
            */
    
            const ramoInicial = criarSVGElemento("line", {
                x1: inicioX,
                y1: 255,
                x2: primeiraEscolhaX - 55,
                y2: yCentral,
                class: "ramo"
            });
    
            ramoInicial.dataset.primeira = pessoa.inicial;
    
            svg.appendChild(ramoInicial);
    
    
            /*
            Nó da primeira pessoa
            */
    
            const rectPessoa = criarSVGElemento("rect", {
                x: primeiraEscolhaX - 55,
                y: yCentral - 16,
                width: 110,
                height: 32,
                rx: 7,
                class: "no-pessoa"
            });
    
            svg.appendChild(rectPessoa);
    
    
            const textoPessoa = criarSVGElemento("text", {
                x: primeiraEscolhaX,
                y: yCentral,
                class: "texto-pessoa"
            });
    
            textoPessoa.textContent = pessoa.nome;
    
            svg.appendChild(textoPessoa);
    
    
            /*
            Segunda escolha
            */
    
            resultadosPessoa.forEach((resultado, indiceLocal) => {
    
                const indiceGlobal =
                    indicePessoa * 3 + indiceLocal;
    
                const yResultado =
                    topo + indiceGlobal * espacamento;
    
    
                /*
                Ramo da primeira escolha até a segunda escolha
                */
    
                const ramo = criarSVGElemento("line", {
                    x1: primeiraEscolhaX + 55,
                    y1: yCentral,
                    x2: segundaEscolhaX - 50,
                    y2: yResultado,
                    class: "ramo"
                });
    
                ramo.dataset.codigo = resultado.codigo;
                ramo.dataset.primeira = pessoa.inicial;
    
                svg.appendChild(ramo);
    
    
                /*
                Caixa com o nome da segunda pessoa
                */
    
                const rectSegundaEscolha = criarSVGElemento("rect", {
                    x: segundaEscolhaX - 50,
                    y: yResultado - 14,
                    width: 100,
                    height: 28,
                    rx: 6,
                    class: "no-pessoa"
                });
    
                svg.appendChild(rectSegundaEscolha);
    
    
                const textoSegundaEscolha = criarSVGElemento("text", {
                    x: segundaEscolhaX,
                    y: yResultado,
                    class: "texto-resultado"
                });
    
                textoSegundaEscolha.textContent =
                    resultado.segunda.nome;
    
                svg.appendChild(textoSegundaEscolha);
    
    
                /*
                Resultado final
                */
    
                const rectResultadoFinal = criarSVGElemento("rect", {
                    x: resultadoFinalX - 27,
                    y: yResultado - 14,
                    width: 54,
                    height: 28,
                    rx: 6,
                    class: "resultado"
                });
    
                rectResultadoFinal.dataset.codigo =
                    resultado.codigo;
    
                rectResultadoFinal.dataset.dupla =
                    resultado.dupla;
    
                svg.appendChild(rectResultadoFinal);
    
    
                const textoResultadoFinal = criarSVGElemento("text", {
                    x: resultadoFinalX,
                    y: yResultado,
                    class: "texto-resultado"
                });
    
                textoResultadoFinal.textContent =
                    resultado.codigo;
    
                svg.appendChild(textoResultadoFinal);
    
    
                /*
                Interação aplicada ao resultado final
                */
    
                if (tipo === "cargos") {
    
                    configurarInteracaoCargos(
                        svg,
                        rectResultadoFinal,
                        resultado
                    );
    
                }
    
                if (tipo === "duplas") {
    
                    configurarInteracaoDuplas(
                        svg,
                        rectResultadoFinal,
                        resultado
                    );
    
                }
    
            });
    
        });
    
    
        /*
        Início
        */
    
        const inicio = criarSVGElemento("circle", {
            cx: inicioX,
            cy: 255,
            r: 8,
            fill: "#555"
        });
    
        svg.appendChild(inicio);
    
    }
    /* ========================================================= */
    /* ÁRVORE 1 — PRESIDENTE E VICE                              */
    /* ========================================================= */

    const svgCargos =
        document.getElementById("arvore-cargos-cap4");

    const resultadoCargos =
        document.getElementById("resultado-cargos-cap4");


    function configurarInteracaoCargos(
        svg,
        elemento,
        resultado
    ) {

        elemento.addEventListener("mouseenter", () => {

            destacarCaminho(
                svg,
                resultado.codigo,
                resultado.primeira.inicial
            );

            resultadoCargos.innerHTML =
                "<b>Presidente:</b> " +
                resultado.primeira.nome +
                " &nbsp;&nbsp; | &nbsp;&nbsp; " +
                "<b>Vice-presidente:</b> " +
                resultado.segunda.nome;

        });


        elemento.addEventListener("mouseleave", () => {

            limparDestaques(svg);

            resultadoCargos.innerHTML =
                "Passe o mouse sobre um dos resultados para interpretar o caminho.";

        });

    }



    /* ========================================================= */
    /* ÁRVORE 2 — DUPLAS                                        */
    /* ========================================================= */

    const svgDuplas =
        document.getElementById("arvore-duplas-cap4");

    const feedbackDuplas =
        document.getElementById("feedback-duplas-cap4");

    const instrucaoDuplas =
        document.getElementById("instrucao-duplas-cap4");

    const resumoDuplas =
        document.getElementById("resumo-duplas-cap4");

    const botaoReiniciar =
        document.getElementById("reiniciar-duplas-cap4");


    let primeiraSelecao = null;

    const paresEncontrados = new Set();


    function configurarInteracaoDuplas(
        svg,
        elemento,
        resultado
    ) {

        elemento.addEventListener("click", () => {

            /*
            Se a dupla já foi encontrada,
            apenas destaca os dois caminhos.
            */

            if (paresEncontrados.has(resultado.dupla)) {

                destacarDupla(
                    svg,
                    resultado.dupla
                );

                feedbackDuplas.innerHTML =
                    "<b>Mesma dupla:</b> " +
                    nomesDaDupla(resultado.dupla);

                return;
            }


            /*
            Primeira escolha
            */

            if (!primeiraSelecao) {

                primeiraSelecao = resultado;

                elemento.classList.add(
                    "resultado-selecionado"
                );

                destacarCaminho(
                    svg,
                    resultado.codigo,
                    resultado.primeira.inicial
                );

                instrucaoDuplas.innerHTML =
                    "<b>2.</b> Agora encontre outro caminho que forme <b>a mesma dupla</b>.";

                feedbackDuplas.innerHTML =
                    "Você escolheu o caminho <b>" +
                    resultado.primeira.nome +
                    " → " +
                    resultado.segunda.nome +
                    "</b>.";

                return;
            }


            /*
            Clicou duas vezes no mesmo resultado
            */

            if (
                resultado.codigo ===
                primeiraSelecao.codigo
            ) {

                feedbackDuplas.innerHTML =
                    "Escolha <b>outro caminho</b> da árvore.";

                return;
            }


            /*
            Encontrou a mesma dupla
            */

            if (
                resultado.dupla ===
                primeiraSelecao.dupla
            ) {

                paresEncontrados.add(
                    resultado.dupla
                );

                colorirDupla(
                    svg,
                    resultado.dupla
                );

                feedbackDuplas.innerHTML =
                    "Muito bem! <b>" +
                    primeiraSelecao.primeira.nome +
                    " → " +
                    primeiraSelecao.segunda.nome +
                    "</b> e <b>" +
                    resultado.primeira.nome +
                    " → " +
                    resultado.segunda.nome +
                    "</b> representam a <b>mesma dupla</b>.";

                primeiraSelecao = null;

                limparSelecaoTemporaria(svg);

                atualizarInstrucao();

                return;
            }


            /*
            Duplas diferentes
            */

            feedbackDuplas.innerHTML =
                "Esses dois caminhos formam <b>duplas diferentes</b>. Tente novamente.";

        });

    }



    /* ========================================================= */
    /* FUNÇÕES DE DESTAQUE                                      */
    /* ========================================================= */

    function destacarCaminho(
        svg,
        codigo,
        primeira
    ) {

        limparDestaques(svg);

        svg.querySelectorAll(".ramo").forEach(ramo => {

            if (
                ramo.dataset.codigo === codigo ||
                (
                    !ramo.dataset.codigo &&
                    ramo.dataset.primeira === primeira
                )
            ) {
                ramo.classList.add(
                    "ramo-destaque"
                );
            }

        });

    }


    function limparDestaques(svg) {

        svg.querySelectorAll(".ramo").forEach(ramo => {
            ramo.classList.remove(
                "ramo-destaque"
            );
        });

    }


    function destacarDupla(svg, dupla) {

        limparDestaques(svg);

        const elementos =
            svg.querySelectorAll(
                '.resultado[data-dupla="' +
                dupla +
                '"]'
            );

        elementos.forEach(el => {

            const codigo =
                el.dataset.codigo;

            const primeira =
                codigo.charAt(0);

            svg.querySelectorAll(".ramo").forEach(
                ramo => {

                    if (
                        ramo.dataset.codigo === codigo ||
                        (
                            !ramo.dataset.codigo &&
                            ramo.dataset.primeira === primeira
                        )
                    ) {

                        ramo.classList.add(
                            "ramo-destaque"
                        );

                    }

                }
            );

        });

    }



    /* ========================================================= */
    /* CORES DAS DUPLAS                                         */
    /* ========================================================= */

    function colorirDupla(svg, dupla) {

        const cor =
            coresDuplas[dupla];

        svg.querySelectorAll(
            '.resultado[data-dupla="' +
            dupla +
            '"]'
        ).forEach(el => {

            el.style.fill = cor;
            el.style.stroke = "#555";
            el.style.strokeWidth = "2.5";

        });

    }


    function limparSelecaoTemporaria(svg) {

        svg.querySelectorAll(
            ".resultado-selecionado"
        ).forEach(el => {

            el.classList.remove(
                "resultado-selecionado"
            );

        });

        limparDestaques(svg);

    }



    /* ========================================================= */
    /* NOMES                                                     */
    /* ========================================================= */

    function nomesDaDupla(dupla) {

        const nomes = dupla
            .split("")
            .map(inicial =>
                pessoas.find(
                    p => p.inicial === inicial
                ).nome
            );

        return nomes.join(" e ");

    }



    /* ========================================================= */
    /* PROGRESSO                                                 */
    /* ========================================================= */

    function atualizarInstrucao() {

        const quantidade =
            paresEncontrados.size;

        if (quantidade < 6) {

            instrucaoDuplas.innerHTML =
                "<b>" +
                quantidade +
                " de 6 duplas identificadas.</b> " +
                "Clique em outro resultado para continuar.";

        } else {

            instrucaoDuplas.innerHTML =
                "<b>Você encontrou todas as duplas!</b>";

            feedbackDuplas.innerHTML =
                "A árvore possui <b>12 caminhos</b>, " +
                "mas eles representam apenas <b>6 duplas diferentes</b>.";

            mostrarResumo();

        }

    }



    function mostrarResumo() {

        resumoDuplas.style.display =
            "block";

        resumoDuplas.innerHTML = `
            <div style="font-weight:bold; margin-bottom:10px;">
                Duplas diferentes
            </div>

            <div style="
                display:flex;
                flex-wrap:wrap;
                gap:8px;
            ">
                ${Object.keys(coresDuplas).map(dupla => `
                    <span style="
                        padding:6px 10px;
                        border-radius:6px;
                        background:${coresDuplas[dupla]};
                        border:1px solid #bbb;
                    ">
                        ${nomesDaDupla(dupla)}
                    </span>
                `).join("")}
            </div>

            <p style="margin-bottom:0; margin-top:14px;">
                <b>Observe:</b> cada dupla apareceu em
                <b>dois caminhos diferentes</b> da árvore.
            </p>
        `;

    }



    /* ========================================================= */
    /* REINICIAR                                                 */
    /* ========================================================= */

    botaoReiniciar.addEventListener(
        "click",
        () => {

            primeiraSelecao = null;

            paresEncontrados.clear();

            feedbackDuplas.innerHTML =
                "Escolha um resultado para começar.";

            instrucaoDuplas.innerHTML =
                "<b>1.</b> Clique em um resultado da árvore.";

            resumoDuplas.style.display =
                "none";

            resumoDuplas.innerHTML = "";

            desenharArvore(
                svgDuplas,
                "duplas"
            );

        }
    );



    /* ========================================================= */
    /* INICIALIZAÇÃO                                             */
    /* ========================================================= */

    desenharArvore(
        svgCargos,
        "cargos"
    );

    desenharArvore(
        svgDuplas,
        "duplas"
    );

})();
</script>
````

Na primeira situação, trocar a ordem das pessoas escolhidas produz um novo resultado: Ana como presidente e Bruno como vice é diferente de Bruno como presidente e Ana como vice.

Na formação de duplas, isso não acontece. Os caminhos Ana → Bruno e Bruno → Ana aparecem separadamente na árvore, mas representam a mesma dupla.

Assim, a árvore mostra 12 caminhos, embora existam apenas 6 duplas diferentes. Isso acontece porque cada dupla foi contada duas vezes: uma para cada ordem possível de escolha de seus integrantes.

## E se quisermos formar grupos de três pessoas?

Agora queremos escolher **3 pessoas entre Ana, Bruno, Caio e Davi** para formar um grupo.

Antes de pensar se a ordem importa ou não, podemos usar a mesma estratégia de decisões sucessivas estudada anteriormente.

* Para a primeira escolha, temos **4 possibilidades**.
* Para a segunda escolha, restam **3 possibilidades**.
* Para a terceira escolha, restam **2 possibilidades**.

Assim, uma árvore completa teria:

$4 \cdot 3 \cdot 2 = 24$

caminhos diferentes.

Mas precisamos interpretar esse resultado com cuidado.

> **Esses 24 caminhos representam realmente 24 grupos diferentes?**

Vamos observar o que acontece com um grupo específico: Ana, Bruno e Caio.

$24 \div 6 = 4$

> **De modo geral, quando escolhemos elementos e a ordem não importa, precisamos identificar quantas vezes cada grupo foi contado apenas por causa das diferentes ordens possíveis.**


```{admonition} Importante
:class: important

Se escolhemos $p$ elementos para formar um grupo, cada grupo pode aparecer em

$p!$
ordens diferentes.

Por isso, quando a ordem **não importa**, precisamos dividir a contagem feita por etapas sucessivas por \(p!\).

### Síntese

Entre Ana, Bruno, Caio e Davi, queremos escolher 3 pessoas para formar um grupo.

Se fizermos a contagem por etapas sucessivas:

- para a primeira escolha, temos 4 possibilidades;
- para a segunda escolha, restam 3 possibilidades;
- para a terceira escolha, restam 2 possibilidades.

Assim,

$$
4 \cdot 3 \cdot 2 = 24
$$

Mas cada grupo de 3 pessoas foi contado várias vezes, pois seus integrantes podem aparecer em diferentes ordens.

O número de ordens possíveis para 3 pessoas é

$$
3! = 3 \cdot 2 \cdot 1 = 6
$$

Portanto,

$$
\frac{4 \cdot 3 \cdot 2}{3!}
=
\frac{24}{6}
=
4
$$

Logo, podemos formar **4 grupos diferentes de três pessoas**.

\_\_

$$

:class: tip, dropdown

```{admonition} Estratégia importante
:class: important

Quando diferentes ordens das mesmas escolhas representam o mesmo resultado:

1. **conte primeiro todos os caminhos possíveis**, como em um processo de escolhas sucessivas;

2. depois, identifique **quantas vezes cada resultado foi contado** apenas por causa das diferentes ordens possíveis;

3. **divida a contagem total por esse número de repetições**.

Assim, eliminamos as contagens repetidas e encontramos apenas os resultados realmente diferentes.

## Quando a ordem não altera o resultado

Nos problemas anteriores, escolhemos alguns elementos de um conjunto e observamos que diferentes ordens dos mesmos elementos podem representar o mesmo resultado.

Na linguagem tradicional da Análise Combinatória, a contagem das maneiras de escolher $p$ elementos entre $n$ elementos distintos, quando a ordem não altera o resultado e não há repetição, recebe o nome de **combinação simples**.

```{admonition} Para saber mais: combinação simples
:class: dropdown

Uma combinação simples é uma escolha de $p$ elementos entre $n$ elementos distintos, na qual diferentes ordens dos mesmos elementos representam o mesmo grupo.

Usando a notação $C_{n,p}$ para o número de combinações de $n$ elementos tomados $p$ a $p$,

$$
C_{n,p}=\frac{n!}{p!(n-p)!}.
$$

Essa fórmula apenas abrevia a estratégia utilizada anteriormente: contar as escolhas por etapas sucessivas e dividir pelo número de ordens possíveis dos elementos escolhidos.
```

````python

%%html

<div id="interacao-ana-cap4">
  <p class="ana4-introducao">
    As duas letras <strong>A</strong> foram marcadas com cores diferentes.
    Observe as seis organizações e encontre aquelas que formam o mesmo
    anagrama quando as cores são desconsideradas.
  </p>

  <div class="ana4-legenda" aria-label="Identificação das letras">
    <span class="ana4-letra ana4-azul" aria-label="A azul">A</span>
    <span class="ana4-letra ana4-neutra" aria-label="N">N</span>
    <span class="ana4-letra ana4-verde" aria-label="A verde">A</span>
  </div>

  <div class="ana4-instrucao" id="ana4-instrucao">
    <strong>Como fazer:</strong> clique em uma organização e depois em outra que
    forme o mesmo anagrama, sem considerar as cores.
  </div>

  <div id="ana4-grade" class="ana4-grade" aria-label="Organizações ainda disponíveis"></div>

  <div id="ana4-feedback" class="ana4-feedback" role="status" aria-live="polite">
    Selecione uma organização para começar.
  </div>

  <section id="ana4-pares" class="ana4-pares" aria-label="Pares encontrados" hidden>
    <h4>Pares encontrados</h4>
    <div id="ana4-lista-pares"></div>
  </section>

  <section id="ana4-pergunta" class="ana4-pergunta" hidden>
    <p><strong>Quantas vezes cada anagrama apareceu entre as seis organizações iniciais?</strong></p>
    <div class="ana4-alternativas" role="group" aria-label="Quantidade de vezes">
      <button type="button" data-resposta="1">1 vez</button>
      <button type="button" data-resposta="2">2 vezes</button>
      <button type="button" data-resposta="3">3 vezes</button>
    </div>
    <div id="ana4-feedback-final" class="ana4-feedback-final" role="status" aria-live="polite"></div>
  </section>

  <div class="ana4-acoes">
    <button type="button" id="ana4-reiniciar">Reiniciar</button>
  </div>
</div>

<style>
#interacao-ana-cap4 {
  font-family: system-ui, -apple-system, BlinkMacSystemFont, "Segoe UI", sans-serif;
  max-width: 820px;
  margin: 20px auto;
  padding: 18px;
  color: #222;
  border: 1px solid #d6d9dc;
  border-radius: 10px;
  background: #fafafa;
  box-sizing: border-box;
}

#interacao-ana-cap4 *,
#interacao-ana-cap4 *::before,
#interacao-ana-cap4 *::after {
  box-sizing: border-box;
}

#interacao-ana-cap4 [hidden] {
  display: none !important;
}

#interacao-ana-cap4 .ana4-introducao {
  margin: 0 0 14px;
  line-height: 1.5;
}

#interacao-ana-cap4 .ana4-legenda {
  display: flex;
  justify-content: center;
  align-items: center;
  gap: 8px;
  margin: 14px 0 18px;
}

#interacao-ana-cap4 .ana4-letra {
  display: inline-flex;
  width: 38px;
  height: 42px;
  align-items: center;
  justify-content: center;
  flex: 0 0 auto;
  border: 1px solid #bdc3cc;
  border-radius: 6px;
  font-weight: 650;
  font-size: 19px;
}

#interacao-ana-cap4 .ana4-azul {
  background: #dbeafe;
  color: #1d4ed8;
  border-color: #93c5fd;
}

#interacao-ana-cap4 .ana4-verde {
  background: #dcfce7;
  color: #166534;
  border-color: #86efac;
}

#interacao-ana-cap4 .ana4-neutra {
  background: #f0f2f4;
  color: #222;
  border-color: #cbd0d6;
}

#interacao-ana-cap4 .ana4-instrucao {
  padding: 12px 14px;
  background: #eef5ff;
  border-left: 4px solid #4f7db8;
  border-radius: 8px;
  margin-bottom: 16px;
  line-height: 1.5;
}

#interacao-ana-cap4 .ana4-grade {
  display: grid;
  grid-template-columns: repeat(2, minmax(0, 1fr));
  gap: 10px;
}

#interacao-ana-cap4 .ana4-organizacao {
  display: flex;
  flex-direction: column;
  align-items: center;
  gap: 8px;
  width: 100%;
  min-width: 0;
  padding: 12px 10px;
  border: 1px solid #d6d9dc;
  border-radius: 8px;
  background: #fff;
  color: #222;
  font: inherit;
  cursor: pointer;
  transition: background-color .15s, border-color .15s;
}

#interacao-ana-cap4 .ana4-organizacao:hover {
  background: #f5f8fc;
}

#interacao-ana-cap4 .ana4-organizacao[aria-pressed="true"] {
  border: 2px solid #4f7db8;
  padding: 11px 9px;
  background: #eef5ff;
}

#interacao-ana-cap4 button:focus-visible {
  outline: 3px solid #4f7db8;
  outline-offset: 2px;
}

#interacao-ana-cap4 {
  font-size: 13px;
  color: #555;
}

#interacao-ana-cap4 .ana4-sequencia {
  display: flex;
  justify-content: center;
  gap: 6px;
}

#interacao-ana-cap4 .ana4-feedback {
  margin-top: 14px;
  min-height: 48px;
  padding: 12px 14px;
  background: #fff;
  border: 1px solid #d6d9dc;
  border-radius: 8px;
  line-height: 1.5;
}

#interacao-ana-cap4 .ana4-feedback.ana4-atencao {
  background: #fff7ed;
  border-color: #fdba74;
}

#interacao-ana-cap4 .ana4-pares {
  margin-top: 20px;
}

#interacao-ana-cap4 .ana4-pares h4 {
  margin: 0 0 10px;
  font-size: 16px;
}

#interacao-ana-cap4 #ana4-lista-pares {
  display: grid;
  gap: 10px;
}

#interacao-ana-cap4 .ana4-par {
  border: 1px solid #d6d9dc;
  background: #fff;
  border-radius: 8px;
  padding: 12px;
}

#interacao-ana-cap4 .ana4-par-organizacoes {
  display: grid;
  grid-template-columns: repeat(2, minmax(0, 1fr));
  gap: 10px;
}

#interacao-ana-cap4 .ana4-mini {
  display: flex;
  gap: 6px;
  align-items: center;
  justify-content: center;
  flex-wrap: wrap;
  min-width: 0;
}

#interacao-ana-cap4 .ana4-par-resultado {
  display: flex;
  align-items: center;
  justify-content: center;
  gap: 8px;
  margin-top: 12px;
  font-size: 14px;
}

#interacao-ana-cap4 .ana4-par-resultado .ana4-letra {
  width: 30px;
  height: 34px;
  font-size: 16px;
}

#interacao-ana-cap4 .ana4-pergunta {
  margin-top: 18px;
  border: 1px solid #d6d9dc;
  background: #fff;
  border-radius: 8px;
  padding: 14px;
}

#interacao-ana-cap4 .ana4-pergunta p {
  margin: 0 0 12px;
}

#interacao-ana-cap4 .ana4-alternativas {
  display: flex;
  flex-wrap: wrap;
  gap: 10px;
}

#interacao-ana-cap4 .ana4-alternativas button,
#interacao-ana-cap4 .ana4-acoes button {
  padding: 9px 14px;
  border: 1px solid #b7bdc5;
  border-radius: 7px;
  background: #fff;
  color: #222;
  font: inherit;
  font-size: 14px;
  cursor: pointer;
}

#interacao-ana-cap4 .ana4-alternativas button:hover,
#interacao-ana-cap4 .ana4-acoes button:hover {
  background: #eef2f6;
}

#interacao-ana-cap4 .ana4-feedback-final {
  margin-top: 10px;
  line-height: 1.5;
}

#interacao-ana-cap4 .ana4-feedback-final.ana4-concluido {
  padding: 12px;
  border-radius: 8px;
  border: 1px solid #86efac;
  background: #ecfdf3;
}

#interacao-ana-cap4 .ana4-acoes {
  margin-top: 16px;
  display: flex;
  justify-content: flex-end;
}

@media (max-width: 560px) {
  #interacao-ana-cap4 {
    padding: 14px;
  }

  #interacao-ana-cap4 .ana4-grade,
  #interacao-ana-cap4 .ana4-par-organizacoes {
    grid-template-columns: 1fr;
  }
}
</style>

<script>
(function () {
  const raiz = document.getElementById("interacao-ana-cap4");
  if (!raiz) return;

  const buscar = seletor => raiz.querySelector(seletor);

  // B: A azul; V: A verde; N: letra N.
  // Cada sequência colorida aparece uma única vez.
  const organizacoes = [
    { id: 1, letras: ["B", "N", "V"] },
    { id: 2, letras: ["V", "B", "N"] },
    { id: 3, letras: ["N", "B", "V"] },
    { id: 4, letras: ["V", "N", "B"] },
    { id: 5, letras: ["N", "V", "B"] },
    { id: 6, letras: ["B", "V", "N"] }
  ];

  const grade = buscar("#ana4-grade");
  const feedback = buscar("#ana4-feedback");
  const areaPares = buscar("#ana4-pares");
  const listaPares = buscar("#ana4-lista-pares");
  const pergunta = buscar("#ana4-pergunta");
  const feedbackFinal = buscar("#ana4-feedback-final");
  const instrucao = buscar("#ana4-instrucao");

  let primeira = null;
  let encontrados = new Set();
  let quantidadePares = 0;

  const anagrama = letras =>
    letras.map(letra => letra === "N" ? "N" : "A").join("");

  const nomeLetra = letra =>
    letra === "B" ? "A azul" :
    letra === "V" ? "A verde" : "N";

  function caixaLetra(letra) {
    const caixa = document.createElement("span");

    caixa.className = "ana4-letra " +
      (letra === "B" ? "ana4-azul" :
       letra === "V" ? "ana4-verde" : "ana4-neutra");

    caixa.textContent = letra === "N" ? "N" : "A";
    caixa.setAttribute("aria-label", nomeLetra(letra));

    return caixa;
  }

  function sequenciaVisual(letras) {
    const linha = document.createElement("div");
    linha.className = "ana4-mini";

    letras.forEach(letra =>
      linha.appendChild(caixaLetra(letra))
    );

    return linha;
  }

  function palavraNeutra(palavra) {
    const linha = document.createElement("div");
    linha.className = "ana4-mini";

    palavra.split("").forEach(letra => {
      const caixa = caixaLetra("N");
      caixa.textContent = letra;
      caixa.setAttribute("aria-label", letra);
      linha.appendChild(caixa);
    });

    return linha;
  }

  function mostrarFeedback(texto, atencao = false) {
    feedback.textContent = texto;
    feedback.classList.toggle("ana4-atencao", atencao);
  }

  function desenharGrade() {
    grade.replaceChildren();

    organizacoes
      .filter(o => !encontrados.has(o.id))
      .forEach(o => {
        const botao = document.createElement("button");
        botao.type = "button";
        botao.className = "ana4-organizacao";

        botao.setAttribute(
          "aria-pressed",
          String(primeira === o.id)
        );

        botao.setAttribute(
          "aria-label",
          `Organização ${o.id}: ${o.letras.map(nomeLetra).join(", ")}`
        );

        const sequencia = document.createElement("div");
        sequencia.className = "ana4-sequencia";

        o.letras.forEach(letra =>
          sequencia.appendChild(caixaLetra(letra))
        );

        botao.appendChild(sequencia);
        botao.addEventListener("click", () => selecionar(o));

        grade.appendChild(botao);
      });
  }

  function registrarPar(a, b) {
    quantidadePares++;
    areaPares.hidden = false;

    const bloco = document.createElement("div");
    bloco.className = "ana4-par";

    const linhas = document.createElement("div");
    linhas.className = "ana4-par-organizacoes";

    [a, b].forEach(o => {
      const celula = document.createElement("div");
      celula.className = "ana4-mini";
      celula.appendChild(sequenciaVisual(o.letras));
      linhas.appendChild(celula);
    });

    const resultado = document.createElement("div");
    resultado.className = "ana4-par-resultado";

    const legenda = document.createElement("span");
    legenda.textContent = "Mesmo anagrama: ";

    resultado.append(
      legenda,
      palavraNeutra(anagrama(a.letras))
    );

    bloco.append(linhas, resultado);
    listaPares.appendChild(bloco);
  }

  function selecionar(o) {
    if (encontrados.has(o.id)) return;

    // Primeira organização escolhida.
    if (primeira === null) {
      primeira = o.id;
      desenharGrade();

      mostrarFeedback(
        "Agora selecione outra organização que forme o mesmo anagrama."
      );
      return;
    }

    // Clicar novamente desfaz a seleção.
    if (primeira === o.id) {
      primeira = null;
      desenharGrade();

      mostrarFeedback(
        "Selecione uma organização para começar."
      );
      return;
    }

    const anterior = organizacoes.find(
      item => item.id === primeira
    );

    // Compara as letras, ignorando suas cores.
    if (anagrama(anterior.letras) !== anagrama(o.letras)) {
      mostrarFeedback(
        "Essas organizações formam anagramas diferentes. " +
        "Observe a posição das letras sem considerar as cores.",
        true
      );
      return;
    }

    // Par correto: registra e retira da grade.
    encontrados.add(anterior.id);
    encontrados.add(o.id);

    primeira = null;
    registrarPar(anterior, o);
    desenharGrade();

    if (encontrados.size === organizacoes.length) {
      grade.hidden = true;
      instrucao.hidden = true;

      mostrarFeedback(
        "Você identificou todas as organizações que formam " +
        "o mesmo anagrama. Observe os pares encontrados."
      );

      pergunta.hidden = false;
    } else {
      mostrarFeedback(
        `Par correto! ${quantidadePares} ` +
        (quantidadePares === 1
          ? "par encontrado"
          : "pares encontrados") +
        ". Continue procurando."
      );
    }
  }

  // Pergunta final: somente após encontrar todos os pares.
  buscar(".ana4-alternativas").addEventListener(
    "click",
    evento => {
      const botao = evento.target.closest(
        "button[data-resposta]"
      );

      if (!botao || !pergunta.contains(botao)) return;

      if (botao.dataset.resposta !== "2") {
        feedbackFinal.textContent =
          "Observe os pares encontrados: quantas organizações " +
          "correspondem a cada anagrama?";
        return;
      }

      feedbackFinal.classList.add("ana4-concluido");

      feedbackFinal.innerHTML =
        "Você encontrou <strong>três anagramas diferentes</strong>: " +
        "ANA, AAN e NAA. Cada um foi contado " +
        "<strong>duas vezes</strong>, pois trocar as duas letras A " +
        "não forma um novo anagrama.";

      buscar(".ana4-alternativas").hidden = true;
    }
  );

  // Reinicia toda a investigação.
  buscar("#ana4-reiniciar").addEventListener("click", () => {
    primeira = null;
    encontrados = new Set();
    quantidadePares = 0;

    listaPares.replaceChildren();
    areaPares.hidden = true;
    pergunta.hidden = true;
    grade.hidden = false;
    instrucao.hidden = false;

    buscar(".ana4-alternativas").hidden = false;
    feedbackFinal.textContent = "";
    feedbackFinal.classList.remove("ana4-concluido");

    mostrarFeedback(
      "Selecione uma organização para começar."
    );

    desenharGrade();
  });

  desenharGrade();
})();
</script>
````

Ao considerar as duas letras A como distintas, encontramos $3! = 6$ organizações.

Mas, como observamos na interação, cada anagrama foi contado duas vezes. Isso acontece porque trocar as duas letras A de posição não produz um novo resultado.

Como os dois As podem trocar de posição de $2! = 2$ maneiras, podemos corrigir a contagem dividindo o total por esse valor:

$$
\frac{3!}{2!}=\frac{6}{2}=3
$$

Portanto, podemos formar **3 anagramas diferentes**: ANA, AAN e NAA.

Utilizamos a mesma estratégia da formação de grupos: contamos inicialmente todas as organizações e depois dividimos para corrigir as repetições.

A diferença é que, agora, **apenas as trocas entre letras iguais não produzem novos resultados**.

## Aplicando a estratégia

Na palavra ANA, conseguimos listar todas as organizações e identificar os anagramas que foram contados mais de uma vez.

Agora, considere a palavra **CASA**.

**De quantas maneiras diferentes podemos organizar as letras dessa palavra?**

Tente resolver sem listar todos os anagramas, utilizando a estratégia que acabamos de estudar.

:::{hint} Dica
:class: dropdown

Se as quatro letras fossem distintas, quantas organizações seriam possíveis? Quais trocas não produzem novos anagramas?
:::

:::{tip} Solução
:class: dropdown

Se as quatro letras fossem distintas, teríamos:

$$
4! = 24
$$

Entretanto, as duas letras A podem trocar de posição sem produzir um novo anagrama. Assim, cada anagrama foi contado $2! = 2$ vezes.

Para corrigir a contagem, dividimos:

$$
\frac{4!}{2!}=\frac{24}{2}=12
$$

Portanto, podemos formar **12 anagramas diferentes** com as letras da palavra CASA.
:::

## Quando há mais de um grupo de letras iguais

Na palavra **OSSO**, temos duas letras O e duas letras S.

**De quantas maneiras diferentes podemos organizar as letras dessa palavra?**

Se considerarmos todas as letras como distintas, teremos:

$$
4!=24
$$

Mas algumas dessas organizações representam o mesmo anagrama.

Pense, por exemplo, no anagrama **OSSO**:

- Quantas trocas podemos fazer entre as duas letras O sem alterar o anagrama?
- E entre as duas letras S?
- Considerando as trocas dos dois pares de letras, quantas vezes esse mesmo anagrama foi contado?

:::{figure} imagens/osso.svg
:width: 90%
:align: center
:alt: Quatro organizações das letras de OSSO, distinguindo as letras iguais por cores.
:::

Como as duas letras O podem ser organizadas de $2!$ maneiras e, para cada uma delas, as duas letras S também podem ser organizadas de $2!$ maneiras, cada anagrama foi contado:

$$
2!\cdot 2!=4
$$

vezes.

Para corrigir essas repetições, dividimos o total de organizações por essa quantidade:

$$
\frac{4!}{2!\cdot 2!}=\frac{24}{4}=6
$$

Portanto, podemos formar **6 anagramas diferentes** com as letras da palavra OSSO.

## Agora é com você

Considere a palavra **ARARA**, formada por três letras A e duas letras R.

**Quantos anagramas diferentes podem ser formados com as letras dessa palavra?**

Utilize as estratégias que estudamos para organizar a contagem.

:::{hint} Dica
:class: dropdown

Comece contando as organizações como se todas as letras fossem distintas. Depois, identifique quantas vezes cada anagrama foi contado devido às trocas entre letras iguais.
:::

:::{tip} Solução
:class: dropdown

Se as cinco letras fossem distintas, teríamos:

$$
5!=120
$$

Entretanto, as três letras A podem trocar de posição de $3!$ maneiras sem produzir novos anagramas. O mesmo acontece com as duas letras R, que podem trocar de posição de $2!$ maneiras.

Pelo Princípio Fundamental da Contagem, cada anagrama foi contado $3!\cdot2!=12$ vezes.

Portanto:

$$
\frac{5!}{3!\cdot2!}=\frac{120}{12}=10
$$

Podemos formar **10 anagramas diferentes** com as letras da palavra ARARA.
:::

Nos exemplos anteriores, contamos inicialmente todas as organizações como se os elementos fossem distintos e depois dividimos para corrigir as repetições.

Essa estratégia também pode ser representada por uma fórmula, utilizando a linguagem tradicional da Análise Combinatória.

:::{seealso} Para saber mais — Permutação com elementos repetidos
:class: dropdown

Na linguagem tradicional da Análise Combinatória, a organização de todos os elementos disponíveis, quando alguns deles são iguais, recebe o nome de **permutação com elementos repetidos**.

Quando temos $n$ elementos, dos quais $\alpha$ são iguais entre si, $\beta$ são iguais entre si, e assim por diante, podemos representar o número de organizações diferentes por:

$$
P_n^{\alpha,\beta,\ldots}=\frac{n!}{\alpha!\cdot\beta!\cdot\ldots}
$$

Nessa expressão:

- $n$ representa o número total de elementos;
- $\alpha$, $\beta$, ... representam as quantidades de elementos iguais de cada tipo;
- $n!$ corresponde à contagem inicial, considerando todos os elementos como distintos;
- os fatoriais do denominador corrigem as repetições produzidas pelas trocas entre elementos iguais.

Por exemplo, na palavra **ARARA**, temos cinco letras, sendo três letras A e duas letras R. Assim:

$$
P_5^{3,2}=\frac{5!}{3!\cdot2!}=10
$$

A fórmula sintetiza a estratégia que utilizamos: contar inicialmente todas as organizações como se os elementos fossem distintos e depois dividir para corrigir as contagens repetidas.

:::
