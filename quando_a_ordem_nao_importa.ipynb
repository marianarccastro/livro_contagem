# Quando a ordem não importa?

No capítulo anterior, vimos situações em que os elementos escolhidos ocupavam posições ou funções específicas. Nesses casos, trocar os elementos de posição produzia um novo resultado.

Agora, vamos comparar duas situações que, à primeira vista, parecem muito parecidas.

Explore os diagramas de árvore. Em cada situação, observe o significado dos caminhos e pense: trocar a ordem das pessoas escolhidas produz um resultado diferente?

```python
%%html

<div id="interacao-ordem-cap4" class="interacao-ordem-cap4">

    <section class="cartao-cap4" aria-labelledby="titulo-cargos-cap4">
        <h3 id="titulo-cargos-cap4">Situação 1 — Presidente e vice-presidente</h3>

        <p>
            Quatro estudantes — <strong>Ana, Bruno, Caio e Davi</strong> — participam
            de uma eleição. Um deles será escolhido presidente e outro vice-presidente.
        </p>

        <p class="instrucao-cap4">
            <strong>Selecione um resultado da árvore</strong> e observe quem ocupa cada função.
        </p>

        <div class="arvore-container-cap4">
            <svg
                id="arvore-cargos-cap4"
                viewBox="0 0 510 520"
                role="group"
                aria-label="Diagrama de árvore para a escolha de presidente e vice-presidente"
            ></svg>
        </div>

        <div
            id="resultado-cargos-cap4"
            class="feedback-cap4"
            aria-live="polite"
        >
            Selecione um resultado para interpretar o caminho.
        </div>
    </section>


    <section class="cartao-cap4" aria-labelledby="titulo-duplas-cap4">
        <h3 id="titulo-duplas-cap4">Situação 2 — Formação de uma dupla</h3>

        <p>
            Agora queremos escolher <strong>duas das mesmas quatro pessoas</strong>
            para formar uma dupla.
        </p>

        <div id="instrucao-duplas-cap4" class="instrucao-cap4" aria-live="polite">
            <strong>1.</strong> Selecione um resultado da árvore.
        </div>

        <div class="arvore-container-cap4">
            <svg
                id="arvore-duplas-cap4"
                viewBox="0 0 510 520"
                role="group"
                aria-label="Diagrama de árvore para a formação de duplas"
            ></svg>
        </div>

        <div
            id="feedback-duplas-cap4"
            class="feedback-cap4"
            aria-live="polite"
        >
            Selecione um resultado para começar.
        </div>

        <div id="resumo-duplas-cap4" class="resumo-cap4" hidden></div>

        <button id="reiniciar-duplas-cap4" class="botao-cap4" type="button">
            Reiniciar
        </button>
    </section>

</div>


<style>

.interacao-ordem-cap4 {
    font-family: system-ui, -apple-system, BlinkMacSystemFont, "Segoe UI", sans-serif;
    max-width: 900px;
    margin: 20px auto;
    color: #222;
}

.interacao-ordem-cap4 .cartao-cap4 {
    border: 1px solid #d6d9dc;
    border-radius: 10px;
    padding: 18px;
    margin-bottom: 28px;
    background: #fafafa;
}

.interacao-ordem-cap4 .cartao-cap4:last-child {
    margin-bottom: 0;
}

.interacao-ordem-cap4 h3 {
    margin-top: 0;
}

.interacao-ordem-cap4 .instrucao-cap4 {
    padding: 12px 14px;
    margin: 14px 0;
    border-left: 4px solid #4f7db8;
    border-radius: 8px;
    background: #eef5ff;
}

.interacao-ordem-cap4 .arvore-container-cap4 {
    width: 100%;
}

.interacao-ordem-cap4 svg {
    display: block;
    width: 100%;
    max-width: 510px;
    height: 520px;
    margin: 0 auto;
}

.interacao-ordem-cap4 .rotulo-etapa-cap4 {
    fill: #333;
    font-size: 14px;
    font-weight: 650;
    text-anchor: middle;
}

.interacao-ordem-cap4 .ramo-cap4 {
    fill: none;
    stroke: #adb5bd;
    stroke-width: 2;
    transition: stroke 0.2s ease, stroke-width 0.2s ease;
}

.interacao-ordem-cap4 .ramo-destaque-cap4 {
    stroke: #333;
    stroke-width: 4;
}

.interacao-ordem-cap4 .no-pessoa-cap4 {
    fill: #fff;
    stroke: #777;
    stroke-width: 1.5;
}

.interacao-ordem-cap4 .texto-pessoa-cap4,
.interacao-ordem-cap4 .texto-resultado-cap4 {
    dominant-baseline: middle;
    text-anchor: middle;
    pointer-events: none;
}

.interacao-ordem-cap4 .texto-pessoa-cap4 {
    font-size: 14px;
}

.interacao-ordem-cap4 .texto-resultado-cap4 {
    font-size: 14px;
    font-weight: 600;
}

.interacao-ordem-cap4 .resultado-cap4 {
    fill: #fff;
    stroke: #777;
    stroke-width: 1.5;
    cursor: pointer;
    transition: fill 0.2s ease, stroke 0.2s ease, stroke-width 0.2s ease;
}

.interacao-ordem-cap4 .resultado-cap4:hover,
.interacao-ordem-cap4 .resultado-cap4:focus {
    stroke: #1f2937;
    stroke-width: 3;
    outline: none;
}

.interacao-ordem-cap4 .resultado-selecionado-cap4 {
    stroke: #1f2937;
    stroke-width: 4;
}

.interacao-ordem-cap4 .feedback-cap4,
.interacao-ordem-cap4 .resumo-cap4 {
    min-height: 54px;
    margin-top: 12px;
    padding: 12px 14px;
    border: 1px solid #ddd;
    border-radius: 8px;
    background: #fff;
}

.interacao-ordem-cap4 .resumo-cap4 {
    margin-top: 16px;
}

.interacao-ordem-cap4 .lista-duplas-cap4 {
    display: flex;
    flex-wrap: wrap;
    gap: 8px;
}

.interacao-ordem-cap4 .dupla-cap4 {
    padding: 6px 10px;
    border: 1px solid #bbb;
    border-radius: 6px;
}

.interacao-ordem-cap4 .botao-cap4 {
    margin-top: 16px;
    padding: 9px 14px;
    border: 1px solid #6b7280;
    border-radius: 7px;
    background: #fff;
    color: #222;
    cursor: pointer;
    font-size: 14px;
}

.interacao-ordem-cap4 .botao-cap4:hover,
.interacao-ordem-cap4 .botao-cap4:focus {
    background: #f3f4f6;
    outline: 2px solid #4f7db8;
    outline-offset: 2px;
}

@media (max-width: 700px) {
    .interacao-ordem-cap4 .cartao-cap4 {
        padding: 14px;
    }
}

</style>


<script>
(function () {
    "use strict";

    const raiz = document.getElementById("interacao-ordem-cap4");

    if (!raiz) {
        return;
    }

    const pessoas = [
        { nome: "Ana", inicial: "A" },
        { nome: "Bruno", inicial: "B" },
        { nome: "Caio", inicial: "C" },
        { nome: "Davi", inicial: "D" }
    ];

    const NS = "http://www.w3.org/2000/svg";

    const coresDuplas = {
        AB: "#ffd6a5",
        AC: "#caffbf",
        AD: "#bde0fe",
        BC: "#ffc8dd",
        BD: "#d8c4f1",
        CD: "#fff3b0"
    };

    const svgCargos = raiz.querySelector("#arvore-cargos-cap4");
    const svgDuplas = raiz.querySelector("#arvore-duplas-cap4");
    const resultadoCargos = raiz.querySelector("#resultado-cargos-cap4");
    const instrucaoDuplas = raiz.querySelector("#instrucao-duplas-cap4");
    const feedbackDuplas = raiz.querySelector("#feedback-duplas-cap4");
    const resumoDuplas = raiz.querySelector("#resumo-duplas-cap4");
    const botaoReiniciar = raiz.querySelector("#reiniciar-duplas-cap4");

    let primeiraSelecao = null;
    let ultimoCargoSelecionado = null;
    let ultimaDuplaDestacada = null;
    const duplasIdentificadas = new Set();


    function criarSVGElemento(tipo, atributos = {}) {
        const elemento = document.createElementNS(NS, tipo);

        Object.entries(atributos).forEach(([chave, valor]) => {
            elemento.setAttribute(chave, valor);
        });

        return elemento;
    }


    function chaveDupla(a, b) {
        return [a, b].sort().join("");
    }


    function nomesDaDupla(dupla) {
        return dupla
            .split("")
            .map(inicial => pessoas.find(pessoa => pessoa.inicial === inicial).nome)
            .join(" e ");
    }


    function gerarResultados() {
        const resultados = [];

        pessoas.forEach(primeira => {
            pessoas.forEach(segunda => {
                if (primeira.inicial !== segunda.inicial) {
                    resultados.push({
                        primeira,
                        segunda,
                        codigo: primeira.inicial + segunda.inicial,
                        dupla: chaveDupla(primeira.inicial, segunda.inicial)
                    });
                }
            });
        });

        return resultados;
    }


    const resultados = gerarResultados();


    function adicionarRotulo(svg, x, texto) {
        const rotulo = criarSVGElemento("text", {
            x,
            y: 27,
            class: "rotulo-etapa-cap4"
        });

        rotulo.textContent = texto;
        svg.appendChild(rotulo);
    }


    function configurarAtivacao(elemento, acao) {
        elemento.addEventListener("click", acao);

        elemento.addEventListener("keydown", evento => {
            if (evento.key === "Enter" || evento.key === " ") {
                evento.preventDefault();
                acao();
            }
        });
    }


    function desenharArvore(svg, tipo) {
        svg.innerHTML = "";

        // A largura do desenho acompanha a largura exibida. Assim, uma
        // unidade do SVG corresponde a um pixel e a fonte não encolhe.
        const largura = svg.getBoundingClientRect().width || 510;
        svg.setAttribute("viewBox", `0 0 ${largura} 520`);
        svg.dataset.larguraDesenhada = String(largura);

        const compacto = largura < 390;
        const muitoEstreito = largura < 285;
        const larguraPrimeira = muitoEstreito ? 72 : compacto ? 78 : 104;
        const larguraSegunda = muitoEstreito ? 72 : compacto ? 78 : 98;
        const larguraResultado = muitoEstreito ? 44 : compacto ? 48 : 56;
        const espacoInicial = muitoEstreito ? 10 : compacto ? 18 : 60;
        const espacoEntreEscolhas = muitoEstreito ? 16 : compacto ? 22 : 85;
        const espacoResultado = muitoEstreito ? 6 : compacto ? 8 : 16;
        const larguraTotal = 16 + espacoInicial + larguraPrimeira +
            espacoEntreEscolhas + larguraSegunda + espacoResultado + larguraResultado;
        const margem = Math.max(0, (largura - larguraTotal) / 2);

        const inicioX = margem + 8;
        const primeiraEsquerda = margem + 16 + espacoInicial;
        const primeiraEscolhaX = primeiraEsquerda + larguraPrimeira / 2;
        const segundaEsquerda = primeiraEsquerda + larguraPrimeira + espacoEntreEscolhas;
        const segundaEscolhaX = segundaEsquerda + larguraSegunda / 2;
        const resultadoEsquerda = segundaEsquerda + larguraSegunda + espacoResultado;
        const resultadoFinalX = resultadoEsquerda + larguraResultado / 2;
        const topo = 70;
        const espacamento = 37;
        const inicioY = topo + 5.5 * espacamento;

        const rotulos = tipo === "cargos"
            ? ["Presidente", "Vice-presidente", "Resultado"]
            : ["1ª escolha", "2ª escolha", "Caminho"];

        adicionarRotulo(svg, primeiraEscolhaX, rotulos[0]);
        adicionarRotulo(svg, segundaEscolhaX, rotulos[1]);
        // Em telas estreitas, deslocar o último rótulo evita que ele
        // encoste no rótulo da segunda escolha.
        const rotuloResultado = criarSVGElemento("text", {
            x: resultadoFinalX,
            y: compacto ? 49 : 27,
            class: "rotulo-etapa-cap4"
        });
        rotuloResultado.textContent = rotulos[2];
        svg.appendChild(rotuloResultado);

        pessoas.forEach((pessoa, indicePessoa) => {
            const resultadosPessoa = resultados.filter(
                resultado => resultado.primeira.inicial === pessoa.inicial
            );

            const primeiroIndice = indicePessoa * 3;
            const yCentral = topo + (primeiroIndice + 1) * espacamento;

            const ramoInicial = criarSVGElemento("line", {
                x1: inicioX + 8,
                y1: inicioY,
                x2: primeiraEsquerda,
                y2: yCentral,
                class: "ramo-cap4"
            });

            ramoInicial.dataset.primeira = pessoa.inicial;
            svg.appendChild(ramoInicial);

            const caixaPrimeiraEscolha = criarSVGElemento("rect", {
                x: primeiraEsquerda,
                y: yCentral - 16,
                width: larguraPrimeira,
                height: 32,
                rx: 7,
                class: "no-pessoa-cap4"
            });

            svg.appendChild(caixaPrimeiraEscolha);

            const textoPrimeiraEscolha = criarSVGElemento("text", {
                x: primeiraEscolhaX,
                y: yCentral,
                class: "texto-pessoa-cap4"
            });

            textoPrimeiraEscolha.textContent = pessoa.nome;
            svg.appendChild(textoPrimeiraEscolha);

            resultadosPessoa.forEach((resultado, indiceLocal) => {
                const indiceGlobal = indicePessoa * 3 + indiceLocal;
                const yResultado = topo + indiceGlobal * espacamento;

                const ramoSegundaEscolha = criarSVGElemento("line", {
                    x1: primeiraEsquerda + larguraPrimeira,
                    y1: yCentral,
                    x2: segundaEsquerda,
                    y2: yResultado,
                    class: "ramo-cap4"
                });

                ramoSegundaEscolha.dataset.codigo = resultado.codigo;
                svg.appendChild(ramoSegundaEscolha);

                const caixaSegundaEscolha = criarSVGElemento("rect", {
                    x: segundaEsquerda,
                    y: yResultado - 14,
                    width: larguraSegunda,
                    height: 28,
                    rx: 6,
                    class: "no-pessoa-cap4"
                });

                svg.appendChild(caixaSegundaEscolha);

                const textoSegundaEscolha = criarSVGElemento("text", {
                    x: segundaEscolhaX,
                    y: yResultado,
                    class: "texto-pessoa-cap4"
                });

                textoSegundaEscolha.textContent = resultado.segunda.nome;
                svg.appendChild(textoSegundaEscolha);

                const caixaResultado = criarSVGElemento("rect", {
                    x: resultadoEsquerda,
                    y: yResultado - 15,
                    width: larguraResultado,
                    height: 30,
                    rx: 6,
                    class: "resultado-cap4",
                    tabindex: "0",
                    role: "button"
                });

                caixaResultado.dataset.codigo = resultado.codigo;
                caixaResultado.dataset.dupla = resultado.dupla;

                const descricao = tipo === "cargos"
                    ? `${resultado.primeira.nome} como presidente e ${resultado.segunda.nome} como vice-presidente`
                    : `Caminho ${resultado.primeira.nome} e depois ${resultado.segunda.nome}`;

                caixaResultado.setAttribute("aria-label", descricao);
                caixaResultado.setAttribute("aria-pressed", "false");
                svg.appendChild(caixaResultado);

                const textoResultado = criarSVGElemento("text", {
                    x: resultadoFinalX,
                    y: yResultado,
                    class: "texto-resultado-cap4"
                });

                textoResultado.textContent = resultado.codigo;
                svg.appendChild(textoResultado);

                if (tipo === "cargos") {
                    configurarInteracaoCargos(svg, caixaResultado, resultado);
                } else {
                    configurarInteracaoDuplas(svg, caixaResultado, resultado);
                }
            });
        });

        const inicio = criarSVGElemento("circle", {
            cx: inicioX,
            cy: inicioY,
            r: 8,
            fill: "#555"
        });

        svg.appendChild(inicio);
    }


    function limparDestaques(svg) {
        svg.querySelectorAll(".ramo-cap4").forEach(ramo => {
            ramo.classList.remove("ramo-destaque-cap4");
        });
    }


    function destacarCaminho(svg, codigo, primeira, limpar = true) {
        if (limpar) {
            limparDestaques(svg);
        }

        svg.querySelectorAll(".ramo-cap4").forEach(ramo => {
            const correspondeAoCodigo = ramo.dataset.codigo === codigo;
            const correspondeAoPrimeiroRamo =
                !ramo.dataset.codigo && ramo.dataset.primeira === primeira;

            if (correspondeAoCodigo || correspondeAoPrimeiroRamo) {
                ramo.classList.add("ramo-destaque-cap4");
            }
        });
    }


    function limparSelecaoVisual(svg) {
        svg.querySelectorAll(".resultado-selecionado-cap4").forEach(elemento => {
            elemento.classList.remove("resultado-selecionado-cap4");
            elemento.setAttribute("aria-pressed", "false");
        });
    }


    function configurarInteracaoCargos(svg, elemento, resultado) {
        configurarAtivacao(elemento, () => {
            ultimoCargoSelecionado = resultado.codigo;
            limparSelecaoVisual(svg);
            limparDestaques(svg);

            elemento.classList.add("resultado-selecionado-cap4");
            elemento.setAttribute("aria-pressed", "true");

            destacarCaminho(
                svg,
                resultado.codigo,
                resultado.primeira.inicial,
                false
            );

            resultadoCargos.innerHTML =
                `<strong>Presidente:</strong> ${resultado.primeira.nome}` +
                ` &nbsp; | &nbsp; ` +
                `<strong>Vice-presidente:</strong> ${resultado.segunda.nome}`;
        });
    }


    function configurarInteracaoDuplas(svg, elemento, resultado) {
        configurarAtivacao(elemento, () => {
            if (duplasIdentificadas.has(resultado.dupla)) {
                ultimaDuplaDestacada = resultado.dupla;
                destacarDupla(svg, resultado.dupla);
                feedbackDuplas.innerHTML =
                    `<strong>Mesma dupla:</strong> ${nomesDaDupla(resultado.dupla)}.`;
                return;
            }

            if (!primeiraSelecao) {
                primeiraSelecao = resultado;
                ultimaDuplaDestacada = null;

                limparSelecaoVisual(svg);
                limparDestaques(svg);
                elemento.classList.add("resultado-selecionado-cap4");
                elemento.setAttribute("aria-pressed", "true");

                destacarCaminho(
                    svg,
                    resultado.codigo,
                    resultado.primeira.inicial,
                    false
                );

                instrucaoDuplas.innerHTML =
                    "<strong>2.</strong> Agora selecione outro caminho que forme " +
                    "<strong>a mesma dupla</strong>.";

                feedbackDuplas.innerHTML =
                    `Você selecionou o caminho <strong>${resultado.primeira.nome} → ` +
                    `${resultado.segunda.nome}</strong>.`;

                return;
            }

            if (resultado.codigo === primeiraSelecao.codigo) {
                feedbackDuplas.innerHTML =
                    "Selecione <strong>outro caminho</strong> da árvore.";
                return;
            }

            if (resultado.dupla === primeiraSelecao.dupla) {
                duplasIdentificadas.add(resultado.dupla);
                ultimaDuplaDestacada = resultado.dupla;
                colorirDupla(svg, resultado.dupla);
                destacarDupla(svg, resultado.dupla);

                feedbackDuplas.innerHTML =
                    `Muito bem! <strong>${primeiraSelecao.primeira.nome} → ` +
                    `${primeiraSelecao.segunda.nome}</strong> e ` +
                    `<strong>${resultado.primeira.nome} → ${resultado.segunda.nome}</strong> ` +
                    "representam a <strong>mesma dupla</strong>.";

                primeiraSelecao = null;
                limparSelecaoVisual(svg);
                atualizarInstrucaoDuplas();
                return;
            }

            feedbackDuplas.innerHTML =
                "Esses dois caminhos formam <strong>duplas diferentes</strong>. " +
                "Mantenha o primeiro caminho e tente outro.";
        });
    }


    function destacarDupla(svg, dupla) {
        limparDestaques(svg);

        svg.querySelectorAll(`.resultado-cap4[data-dupla="${dupla}"]`).forEach(elemento => {
            const codigo = elemento.dataset.codigo;

            destacarCaminho(svg, codigo, codigo.charAt(0), false);
        });
    }


    function colorirDupla(svg, dupla) {
        const cor = coresDuplas[dupla];

        svg.querySelectorAll(`.resultado-cap4[data-dupla="${dupla}"]`).forEach(elemento => {
            elemento.style.fill = cor;
            elemento.style.stroke = "#555";
            elemento.style.strokeWidth = "2.5";
        });
    }


    function atualizarInstrucaoDuplas() {
        const quantidade = duplasIdentificadas.size;

        if (quantidade < Object.keys(coresDuplas).length) {
            const palavra = quantidade === 1 ? "dupla identificada" : "duplas identificadas";

            instrucaoDuplas.innerHTML =
                `<strong>${quantidade} ${palavra}.</strong> ` +
                "Selecione outro resultado para continuar.";
            return;
        }

        instrucaoDuplas.innerHTML =
            "<strong>Você identificou todas as duplas!</strong>";

        feedbackDuplas.innerHTML =
            "A árvore possui <strong>12 caminhos</strong>, mas eles representam apenas " +
            "<strong>6 duplas diferentes</strong>.";

        mostrarResumo();
    }


    function mostrarResumo() {
        resumoDuplas.hidden = false;

        const itens = Object.keys(coresDuplas)
            .map(dupla =>
                `<span class="dupla-cap4" style="background:${coresDuplas[dupla]}">` +
                `${nomesDaDupla(dupla)}</span>`
            )
            .join("");

        resumoDuplas.innerHTML = `
            <div style="font-weight:700; margin-bottom:10px;">Duplas diferentes</div>
            <div class="lista-duplas-cap4">${itens}</div>
            <p style="margin-bottom:0; margin-top:14px;">
                <strong>Observe:</strong> cada dupla apareceu em
                <strong>dois caminhos diferentes</strong> da árvore.
            </p>
        `;
    }


    function reiniciarDuplas() {
        primeiraSelecao = null;
        ultimaDuplaDestacada = null;
        duplasIdentificadas.clear();

        instrucaoDuplas.innerHTML =
            "<strong>1.</strong> Selecione um resultado da árvore.";

        feedbackDuplas.innerHTML =
            "Selecione um resultado para começar.";

        resumoDuplas.hidden = true;
        resumoDuplas.innerHTML = "";

        desenharArvore(svgDuplas, "duplas");
    }


    function redesenharPreservandoSelecoes() {
        const larguraAtual = svgCargos.getBoundingClientRect().width;
        const larguraDesenhada = Number(svgCargos.dataset.larguraDesenhada || 0);

        if (Math.abs(larguraAtual - larguraDesenhada) < 1) {
            return;
        }

        const elementoFocado = document.activeElement;
        const codigoFocado = elementoFocado && elementoFocado.dataset
            ? elementoFocado.dataset.codigo
            : null;
        const arvoreFocada = codigoFocado && svgCargos.contains(elementoFocado)
            ? "cargos"
            : codigoFocado && svgDuplas.contains(elementoFocado)
                ? "duplas"
                : null;

        desenharArvore(svgCargos, "cargos");
        desenharArvore(svgDuplas, "duplas");

        if (ultimoCargoSelecionado) {
            const selecionado = svgCargos.querySelector(
                `.resultado-cap4[data-codigo="${ultimoCargoSelecionado}"]`
            );

            if (selecionado) {
                selecionado.classList.add("resultado-selecionado-cap4");
                selecionado.setAttribute("aria-pressed", "true");
                destacarCaminho(svgCargos, ultimoCargoSelecionado, ultimoCargoSelecionado[0]);
            }
        }

        duplasIdentificadas.forEach(dupla => colorirDupla(svgDuplas, dupla));

        if (primeiraSelecao) {
            const selecionado = svgDuplas.querySelector(
                `.resultado-cap4[data-codigo="${primeiraSelecao.codigo}"]`
            );

            if (selecionado) {
                selecionado.classList.add("resultado-selecionado-cap4");
                selecionado.setAttribute("aria-pressed", "true");
                destacarCaminho(
                    svgDuplas,
                    primeiraSelecao.codigo,
                    primeiraSelecao.primeira.inicial
                );
            }
        } else if (ultimaDuplaDestacada) {
            destacarDupla(svgDuplas, ultimaDuplaDestacada);
        }

        if (arvoreFocada) {
            const svg = arvoreFocada === "cargos" ? svgCargos : svgDuplas;
            const novoElemento = svg.querySelector(
                `.resultado-cap4[data-codigo="${codigoFocado}"]`
            );

            if (novoElemento) {
                novoElemento.focus();
            }
        }
    }


    botaoReiniciar.addEventListener("click", reiniciarDuplas);

    desenharArvore(svgCargos, "cargos");
    desenharArvore(svgDuplas, "duplas");

    if (typeof ResizeObserver !== "undefined") {
        const observador = new ResizeObserver(redesenharPreservandoSelecoes);
        observador.observe(svgCargos);
    } else {
        window.addEventListener("resize", redesenharPreservandoSelecoes);
    }
})();
</script>
```

As seis ordens exibidas são as **permutações dos integrantes** do grupo formado por Ana, Bruno e Caio. Como vimos no capítulo anterior, o número de permutações de três pessoas distintas é $3! = 6$.

Cada permutação corresponde a um caminho da contagem por etapas, mas todas formam o **mesmo grupo**. Portanto, nos 24 caminhos contados inicialmente, cada grupo de três pessoas aparece 6 vezes.

Para encontrar o número de grupos diferentes, dividimos a contagem inicial por esse número de repetições:

$$
\frac{4 \cdot 3 \cdot 2}{3!}
=
\frac{24}{6}
=
4
$$

Logo, podemos formar **4 grupos diferentes de três pessoas**.

:::{important} Importante

Se escolhemos $p$ elementos distintos para formar um grupo, cada grupo aparece em $p!$ ordens diferentes na contagem por etapas sucessivas.

Por isso, quando a ordem **não importa**, dividimos essa contagem por $p!$ para encontrar o número de grupos diferentes.

:::

Ao escolher os três estudantes sucessivamente, contamos

6 \cdot 5 \cdot 4 = 120

caminhos. Cada equipe aparece em $3! = 6$ desses caminhos, um para cada permutação de seus integrantes. Portanto, o número de equipes diferentes é

:::{tip} Estratégia importante

Quando vários caminhos da contagem inicial representam o mesmo resultado:

1. **conte primeiro todos os caminhos possíveis**, como em um processo de escolhas sucessivas;

2. depois, identifique **quantas vezes cada resultado foi contado** apenas por causa das diferentes ordens possíveis;

3. se cada resultado foi contado o mesmo número de vezes, **divida a contagem total por essa quantidade**.

Assim, eliminamos as contagens repetidas e encontramos apenas os resultados realmente diferentes.

:::

## Quando a ordem não altera o resultado

Nos problemas anteriores, escolhemos alguns elementos de um conjunto e observamos que diferentes ordens dos mesmos elementos podem representar o mesmo resultado.

Na linguagem tradicional da Análise Combinatória, a contagem das maneiras de escolher $p$ elementos entre $n$ elementos distintos, quando a ordem não altera o resultado e não há repetição, recebe o nome de **combinação simples**.

:::{seealso} Para saber mais: Combinação simples
:class: dropdown

Para escolher $p$ elementos entre $n$ elementos distintos por etapas sucessivas, temos $n$ opções na primeira escolha, $n-1$ na segunda, e assim por diante. Na última das $p$ escolhas, restam $n-p+1$ opções. Portanto, a contagem inicial é

$$
n \cdot (n-1) \cdots (n-p+1)
=
\frac{n!}{(n-p)!}.
$$

Dividimos $n!$ por $(n-p)!$ para retirar os fatores que não fazem parte dessas $p$ escolhas. Como cada grupo aparece em $p!$ ordens diferentes, dividimos também por $p!$.

Chamando de $C_{n,p}$ o número de combinações simples, obtemos

$$
C_{n,p}=\frac{n!}{p!(n-p)!}.
$$

Essa fórmula apenas abrevia a estratégia utilizada anteriormente: contar as escolhas por etapas sucessivas e dividir pelo número de ordens possíveis dos elementos escolhidos.

:::

Nos exemplos anteriores, contamos inicialmente todas as organizações como se os elementos fossem distintos e depois dividimos para corrigir as repetições.

Essa estratégia também pode ser representada por uma fórmula, utilizando a linguagem tradicional da Análise Combinatória.

:::{seealso} Para saber mais — Permutação com elementos repetidos
:class: dropdown

Na linguagem tradicional da Análise Combinatória, a organização de todos os elementos disponíveis, quando alguns deles são iguais, recebe o nome de **permutação com elementos repetidos**.

Quando temos $n$ elementos, dos quais $\alpha$ são iguais entre si, $\beta$ são iguais entre si, e assim por diante, podemos representar o número de organizações diferentes por:

$$
P_n^{\alpha,\beta,\ldots}=\frac{n!}{\alpha!\cdot\beta!\cdot\ldots}
$$

Nessa expressão:

- $n$ representa o número total de elementos;
- $\alpha$, $\beta$, ... representam as quantidades de elementos iguais de cada tipo;
- $n!$ corresponde à contagem inicial, considerando todos os elementos como distintos;
- os fatoriais do denominador corrigem as repetições produzidas pelas trocas entre elementos iguais.

Por exemplo, na palavra **ARARA**, temos cinco letras, sendo três letras A e duas letras R. Assim:

$$
P_5^{3,2}=\frac{5!}{3!\cdot2!}=10
$$

A fórmula sintetiza a estratégia que utilizamos: contar inicialmente todas as organizações como se os elementos fossem distintos e depois dividir para corrigir as contagens repetidas.
