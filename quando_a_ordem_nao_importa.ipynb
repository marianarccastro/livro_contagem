````python
%%html

<div id="interacao-ordem-cap4" style="
    font-family: system-ui, sans-serif;
    max-width: 900px;
    margin: 20px auto;
    color: #222;
">

    <!-- ========================================================= -->
    <!-- ÁRVORE 1 -->
    <!-- ========================================================= -->

    <div style="
        border: 1px solid #d6d9dc;
        border-radius: 10px;
        padding: 18px;
        margin-bottom: 28px;
        background: #fafafa;
    ">
        <h3 style="margin-top:0;">Situação 1 — Presidente e vice-presidente</h3>

        <p>
            Quatro estudantes — <b>Ana, Bruno, Caio e Davi</b> — participam de uma eleição.
            Um deles será escolhido presidente e outro vice-presidente.
        </p>

        <p style="margin-bottom:8px;">
            <b>Passe o mouse sobre um resultado da árvore.</b>
        </p>

        <div style="overflow-x:auto;">
            <svg
                id="arvore-cargos-cap4"
                viewBox="0 0 740 510"
                style="width:100%;"
                aria-label="Diagrama de árvore para escolha de presidente e vice-presidente"
            ></svg>
        </div>

        <div id="resultado-cargos-cap4" style="
            min-height: 54px;
            margin-top: 12px;
            padding: 12px 14px;
            border-radius: 8px;
            background: white;
            border: 1px solid #ddd;
        ">
            Passe o mouse sobre um dos resultados para interpretar o caminho.
        </div>
    </div>


    <!-- ========================================================= -->
    <!-- ÁRVORE 2 -->
    <!-- ========================================================= -->

    <div style="
        border: 1px solid #d6d9dc;
        border-radius: 10px;
        padding: 18px;
        background: #fafafa;
    ">
        <h3 style="margin-top:0;">Situação 2 — Formação de uma dupla</h3>

        <p>
            Agora queremos escolher <b>duas das mesmas quatro pessoas</b>
            para formar uma dupla.
        </p>

        <div id="instrucao-duplas-cap4" style="
            padding: 12px 14px;
            margin: 14px 0;
            border-radius: 8px;
            background: #eef5ff;
            border-left: 4px solid #4f7db8;
        ">
            <b>1.</b> Clique em um resultado da árvore.
        </div>

        <div style="overflow-x:auto;">
            <svg
                id="arvore-duplas-cap4"
                viewBox="0 0 740 510"
                style="width:100%;"
                aria-label="Diagrama de árvore para formação de duplas"
            ></svg>
        </div>

        <div id="feedback-duplas-cap4" style="
            min-height: 54px;
            margin-top: 12px;
            padding: 12px 14px;
            border-radius: 8px;
            background: white;
            border: 1px solid #ddd;
        ">
            Escolha um resultado para começar.
        </div>

        <div id="resumo-duplas-cap4" style="
            display:none;
            margin-top:16px;
            padding:14px;
            border-radius:8px;
            border:1px solid #ccc;
            background:white;
        "></div>

        <button id="reiniciar-duplas-cap4" style="
            margin-top: 16px;
            padding: 9px 16px;
            border: none;
            border-radius: 6px;
            background: #555;
            color: white;
            cursor: pointer;
            font-size: 14px;
        ">
            Reiniciar
        </button>
    </div>

</div>


<style>

#interacao-ordem-cap4 .ramo {
    stroke: #adb5bd;
    stroke-width: 2;
    fill: none;
    transition: all 0.2s ease;
}

#interacao-ordem-cap4 .ramo-destaque {
    stroke: #333;
    stroke-width: 4;
}

#interacao-ordem-cap4 .no-pessoa {
    fill: white;
    stroke: #777;
    stroke-width: 1.5;
}

#interacao-ordem-cap4 .texto-pessoa {
    font-size: 15px;
    dominant-baseline: middle;
    text-anchor: middle;
    pointer-events: none;
}

#interacao-ordem-cap4 .resultado {
    fill: white;
    stroke: #888;
    stroke-width: 1.5;
    cursor: pointer;
    transition: all 0.2s ease;
}

#interacao-ordem-cap4 .resultado:hover {
    stroke-width: 3;
}

#interacao-ordem-cap4 .texto-resultado {
    font-size: 14px;
    dominant-baseline: middle;
    text-anchor: middle;
    pointer-events: none;
}

#interacao-ordem-cap4 .resultado-selecionado {
    stroke: #222;
    stroke-width: 4;
}

</style>


<script>
(function () {

    const pessoas = [
        { nome: "Ana", inicial: "A" },
        { nome: "Bruno", inicial: "B" },
        { nome: "Caio", inicial: "C" },
        { nome: "Davi", inicial: "D" }
    ];

    const NS = "http://www.w3.org/2000/svg";

    const coresDuplas = {
        "AB": "#ffd6a5",
        "AC": "#caffbf",
        "AD": "#bde0fe",
        "BC": "#ffc8dd",
        "BD": "#d8c4f1",
        "CD": "#fff3b0"
    };

    function criarSVGElemento(tipo, atributos = {}) {
        const el = document.createElementNS(NS, tipo);

        Object.entries(atributos).forEach(([chave, valor]) => {
            el.setAttribute(chave, valor);
        });

        return el;
    }


    function chaveDupla(a, b) {
        return [a, b].sort().join("");
    }


    function gerarResultados() {

        const resultados = [];

        pessoas.forEach(primeira => {

            pessoas.forEach(segunda => {

                if (primeira.inicial !== segunda.inicial) {

                    resultados.push({
                        primeira,
                        segunda,
                        codigo: primeira.inicial + segunda.inicial,
                        dupla: chaveDupla(
                            primeira.inicial,
                            segunda.inicial
                        )
                    });

                }

            });

        });

        return resultados;
    }


    const resultados = gerarResultados();


    /* ========================================================= */
    /* FUNÇÃO GERAL PARA DESENHAR ÁRVORE                         */
    /* ========================================================= */

    function desenharArvore(svg, tipo) {
    
        svg.innerHTML = "";
    
        const inicioX = 45;
        const primeiraEscolhaX = 220;
        const segundaEscolhaX = 500;
        const resultadoFinalX = 590;
    
        const topo = 35;
        const espacamento = 38;
    
        /*
        Cada pessoa ocupa um bloco com três resultados.
        O nó da primeira escolha fica no centro desses três.
        */
    
        pessoas.forEach((pessoa, indicePessoa) => {
    
            const resultadosPessoa = resultados.filter(
                r => r.primeira.inicial === pessoa.inicial
            );
    
            const primeiroIndice = indicePessoa * 3;
    
            const yCentral =
                topo +
                (primeiroIndice + 1) * espacamento;
    
    
            /*
            Ramo do início até a primeira escolha
            */
    
            const ramoInicial = criarSVGElemento("line", {
                x1: inicioX,
                y1: 255,
                x2: primeiraEscolhaX - 55,
                y2: yCentral,
                class: "ramo"
            });
    
            ramoInicial.dataset.primeira = pessoa.inicial;
    
            svg.appendChild(ramoInicial);
    
    
            /*
            Nó da primeira pessoa
            */
    
            const rectPessoa = criarSVGElemento("rect", {
                x: primeiraEscolhaX - 55,
                y: yCentral - 16,
                width: 110,
                height: 32,
                rx: 7,
                class: "no-pessoa"
            });
    
            svg.appendChild(rectPessoa);
    
    
            const textoPessoa = criarSVGElemento("text", {
                x: primeiraEscolhaX,
                y: yCentral,
                class: "texto-pessoa"
            });
    
            textoPessoa.textContent = pessoa.nome;
    
            svg.appendChild(textoPessoa);
    
    
            /*
            Segunda escolha
            */
    
            resultadosPessoa.forEach((resultado, indiceLocal) => {
    
                const indiceGlobal =
                    indicePessoa * 3 + indiceLocal;
    
                const yResultado =
                    topo + indiceGlobal * espacamento;
    
    
                /*
                Ramo da primeira escolha até a segunda escolha
                */
    
                const ramo = criarSVGElemento("line", {
                    x1: primeiraEscolhaX + 55,
                    y1: yCentral,
                    x2: segundaEscolhaX - 50,
                    y2: yResultado,
                    class: "ramo"
                });
    
                ramo.dataset.codigo = resultado.codigo;
                ramo.dataset.primeira = pessoa.inicial;
    
                svg.appendChild(ramo);
    
    
                /*
                Caixa com o nome da segunda pessoa
                */
    
                const rectSegundaEscolha = criarSVGElemento("rect", {
                    x: segundaEscolhaX - 50,
                    y: yResultado - 14,
                    width: 100,
                    height: 28,
                    rx: 6,
                    class: "no-pessoa"
                });
    
                svg.appendChild(rectSegundaEscolha);
    
    
                const textoSegundaEscolha = criarSVGElemento("text", {
                    x: segundaEscolhaX,
                    y: yResultado,
                    class: "texto-resultado"
                });
    
                textoSegundaEscolha.textContent =
                    resultado.segunda.nome;
    
                svg.appendChild(textoSegundaEscolha);
    
    
                /*
                Resultado final
                */
    
                const rectResultadoFinal = criarSVGElemento("rect", {
                    x: resultadoFinalX - 27,
                    y: yResultado - 14,
                    width: 54,
                    height: 28,
                    rx: 6,
                    class: "resultado"
                });
    
                rectResultadoFinal.dataset.codigo =
                    resultado.codigo;
    
                rectResultadoFinal.dataset.dupla =
                    resultado.dupla;
    
                svg.appendChild(rectResultadoFinal);
    
    
                const textoResultadoFinal = criarSVGElemento("text", {
                    x: resultadoFinalX,
                    y: yResultado,
                    class: "texto-resultado"
                });
    
                textoResultadoFinal.textContent =
                    resultado.codigo;
    
                svg.appendChild(textoResultadoFinal);
    
    
                /*
                Interação aplicada ao resultado final
                */
    
                if (tipo === "cargos") {
    
                    configurarInteracaoCargos(
                        svg,
                        rectResultadoFinal,
                        resultado
                    );
    
                }
    
                if (tipo === "duplas") {
    
                    configurarInteracaoDuplas(
                        svg,
                        rectResultadoFinal,
                        resultado
                    );
    
                }
    
            });
    
        });
    
    
        /*
        Início
        */
    
        const inicio = criarSVGElemento("circle", {
            cx: inicioX,
            cy: 255,
            r: 8,
            fill: "#555"
        });
    
        svg.appendChild(inicio);
    
    }
    /* ========================================================= */
    /* ÁRVORE 1 — PRESIDENTE E VICE                              */
    /* ========================================================= */

    const svgCargos =
        document.getElementById("arvore-cargos-cap4");

    const resultadoCargos =
        document.getElementById("resultado-cargos-cap4");


    function configurarInteracaoCargos(
        svg,
        elemento,
        resultado
    ) {

        elemento.addEventListener("mouseenter", () => {

            destacarCaminho(
                svg,
                resultado.codigo,
                resultado.primeira.inicial
            );

            resultadoCargos.innerHTML =
                "<b>Presidente:</b> " +
                resultado.primeira.nome +
                " &nbsp;&nbsp; | &nbsp;&nbsp; " +
                "<b>Vice-presidente:</b> " +
                resultado.segunda.nome;

        });


        elemento.addEventListener("mouseleave", () => {

            limparDestaques(svg);

            resultadoCargos.innerHTML =
                "Passe o mouse sobre um dos resultados para interpretar o caminho.";

        });

    }



    /* ========================================================= */
    /* ÁRVORE 2 — DUPLAS                                        */
    /* ========================================================= */

    const svgDuplas =
        document.getElementById("arvore-duplas-cap4");

    const feedbackDuplas =
        document.getElementById("feedback-duplas-cap4");

    const instrucaoDuplas =
        document.getElementById("instrucao-duplas-cap4");

    const resumoDuplas =
        document.getElementById("resumo-duplas-cap4");

    const botaoReiniciar =
        document.getElementById("reiniciar-duplas-cap4");


    let primeiraSelecao = null;

    const paresEncontrados = new Set();


    function configurarInteracaoDuplas(
        svg,
        elemento,
        resultado
    ) {

        elemento.addEventListener("click", () => {

            /*
            Se a dupla já foi encontrada,
            apenas destaca os dois caminhos.
            */

            if (paresEncontrados.has(resultado.dupla)) {

                destacarDupla(
                    svg,
                    resultado.dupla
                );

                feedbackDuplas.innerHTML =
                    "<b>Mesma dupla:</b> " +
                    nomesDaDupla(resultado.dupla);

                return;
            }


            /*
            Primeira escolha
            */

            if (!primeiraSelecao) {

                primeiraSelecao = resultado;

                elemento.classList.add(
                    "resultado-selecionado"
                );

                destacarCaminho(
                    svg,
                    resultado.codigo,
                    resultado.primeira.inicial
                );

                instrucaoDuplas.innerHTML =
                    "<b>2.</b> Agora encontre outro caminho que forme <b>a mesma dupla</b>.";

                feedbackDuplas.innerHTML =
                    "Você escolheu o caminho <b>" +
                    resultado.primeira.nome +
                    " → " +
                    resultado.segunda.nome +
                    "</b>.";

                return;
            }


            /*
            Clicou duas vezes no mesmo resultado
            */

            if (
                resultado.codigo ===
                primeiraSelecao.codigo
            ) {

                feedbackDuplas.innerHTML =
                    "Escolha <b>outro caminho</b> da árvore.";

                return;
            }


            /*
            Encontrou a mesma dupla
            */

            if (
                resultado.dupla ===
                primeiraSelecao.dupla
            ) {

                paresEncontrados.add(
                    resultado.dupla
                );

                colorirDupla(
                    svg,
                    resultado.dupla
                );

                feedbackDuplas.innerHTML =
                    "Muito bem! <b>" +
                    primeiraSelecao.primeira.nome +
                    " → " +
                    primeiraSelecao.segunda.nome +
                    "</b> e <b>" +
                    resultado.primeira.nome +
                    " → " +
                    resultado.segunda.nome +
                    "</b> representam a <b>mesma dupla</b>.";

                primeiraSelecao = null;

                limparSelecaoTemporaria(svg);

                atualizarInstrucao();

                return;
            }


            /*
            Duplas diferentes
            */

            feedbackDuplas.innerHTML =
                "Esses dois caminhos formam <b>duplas diferentes</b>. Tente novamente.";

        });

    }



    /* ========================================================= */
    /* FUNÇÕES DE DESTAQUE                                      */
    /* ========================================================= */

    function destacarCaminho(
        svg,
        codigo,
        primeira
    ) {

        limparDestaques(svg);

        svg.querySelectorAll(".ramo").forEach(ramo => {

            if (
                ramo.dataset.codigo === codigo ||
                (
                    !ramo.dataset.codigo &&
                    ramo.dataset.primeira === primeira
                )
            ) {
                ramo.classList.add(
                    "ramo-destaque"
                );
            }

        });

    }


    function limparDestaques(svg) {

        svg.querySelectorAll(".ramo").forEach(ramo => {
            ramo.classList.remove(
                "ramo-destaque"
            );
        });

    }


    function destacarDupla(svg, dupla) {

        limparDestaques(svg);

        const elementos =
            svg.querySelectorAll(
                '.resultado[data-dupla="' +
                dupla +
                '"]'
            );

        elementos.forEach(el => {

            const codigo =
                el.dataset.codigo;

            const primeira =
                codigo.charAt(0);

            svg.querySelectorAll(".ramo").forEach(
                ramo => {

                    if (
                        ramo.dataset.codigo === codigo ||
                        (
                            !ramo.dataset.codigo &&
                            ramo.dataset.primeira === primeira
                        )
                    ) {

                        ramo.classList.add(
                            "ramo-destaque"
                        );

                    }

                }
            );

        });

    }



    /* ========================================================= */
    /* CORES DAS DUPLAS                                         */
    /* ========================================================= */

    function colorirDupla(svg, dupla) {

        const cor =
            coresDuplas[dupla];

        svg.querySelectorAll(
            '.resultado[data-dupla="' +
            dupla +
            '"]'
        ).forEach(el => {

            el.style.fill = cor;
            el.style.stroke = "#555";
            el.style.strokeWidth = "2.5";

        });

    }


    function limparSelecaoTemporaria(svg) {

        svg.querySelectorAll(
            ".resultado-selecionado"
        ).forEach(el => {

            el.classList.remove(
                "resultado-selecionado"
            );

        });

        limparDestaques(svg);

    }



    /* ========================================================= */
    /* NOMES                                                     */
    /* ========================================================= */

    function nomesDaDupla(dupla) {

        const nomes = dupla
            .split("")
            .map(inicial =>
                pessoas.find(
                    p => p.inicial === inicial
                ).nome
            );

        return nomes.join(" e ");

    }



    /* ========================================================= */
    /* PROGRESSO                                                 */
    /* ========================================================= */

    function atualizarInstrucao() {

        const quantidade =
            paresEncontrados.size;

        if (quantidade < 6) {

            instrucaoDuplas.innerHTML =
                "<b>" +
                quantidade +
                " de 6 duplas identificadas.</b> " +
                "Clique em outro resultado para continuar.";

        } else {

            instrucaoDuplas.innerHTML =
                "<b>Você encontrou todas as duplas!</b>";

            feedbackDuplas.innerHTML =
                "A árvore possui <b>12 caminhos</b>, " +
                "mas eles representam apenas <b>6 duplas diferentes</b>.";

            mostrarResumo();

        }

    }



    function mostrarResumo() {

        resumoDuplas.style.display =
            "block";

        resumoDuplas.innerHTML = `
            <div style="font-weight:bold; margin-bottom:10px;">
                Duplas diferentes
            </div>

            <div style="
                display:flex;
                flex-wrap:wrap;
                gap:8px;
            ">
                ${Object.keys(coresDuplas).map(dupla => `
                    <span style="
                        padding:6px 10px;
                        border-radius:6px;
                        background:${coresDuplas[dupla]};
                        border:1px solid #bbb;
                    ">
                        ${nomesDaDupla(dupla)}
                    </span>
                `).join("")}
            </div>

            <p style="margin-bottom:0; margin-top:14px;">
                <b>Observe:</b> cada dupla apareceu em
                <b>dois caminhos diferentes</b> da árvore.
            </p>
        `;

    }



    /* ========================================================= */
    /* REINICIAR                                                 */
    /* ========================================================= */

    botaoReiniciar.addEventListener(
        "click",
        () => {

            primeiraSelecao = null;

            paresEncontrados.clear();

            feedbackDuplas.innerHTML =
                "Escolha um resultado para começar.";

            instrucaoDuplas.innerHTML =
                "<b>1.</b> Clique em um resultado da árvore.";

            resumoDuplas.style.display =
                "none";

            resumoDuplas.innerHTML = "";

            desenharArvore(
                svgDuplas,
                "duplas"
            );

        }
    );



    /* ========================================================= */
    /* INICIALIZAÇÃO                                             */
    /* ========================================================= */

    desenharArvore(
        svgCargos,
        "cargos"
    );

    desenharArvore(
        svgDuplas,
        "duplas"
    );

})();
</script>
````

Na primeira situação, trocar a ordem das pessoas escolhidas produz um novo resultado: Ana como presidente e Bruno como vice é diferente de Bruno como presidente e Ana como vice.

Na formação de duplas, isso não acontece. Os caminhos Ana → Bruno e Bruno → Ana aparecem separadamente na árvore, mas representam a mesma dupla.

Assim, a árvore mostra 12 caminhos, embora existam apenas 6 duplas diferentes. Isso acontece porque cada dupla foi contada duas vezes: uma para cada ordem possível de escolha de seus integrantes.

## E se quisermos formar grupos de três pessoas?

Agora queremos escolher **3 pessoas entre Ana, Bruno, Caio e Davi** para formar um grupo.

Antes de pensar se a ordem importa ou não, podemos usar a mesma estratégia de decisões sucessivas estudada anteriormente.

* Para a primeira escolha, temos **4 possibilidades**.
* Para a segunda escolha, restam **3 possibilidades**.
* Para a terceira escolha, restam **2 possibilidades**.

Assim, uma árvore completa teria:

$4 \cdot 3 \cdot 2 = 24$

caminhos diferentes.

Mas precisamos interpretar esse resultado com cuidado.

> **Esses 24 caminhos representam realmente 24 grupos diferentes?**

Vamos observar o que acontece com um grupo específico: Ana, Bruno e Caio.

O grupo formado por **Ana, Bruno e Caio** apareceu em **6 caminhos diferentes**:

$3 \cdot 2 \cdot 1 = 6$

Isso acontece porque, embora a ordem das escolhas mude, o grupo formado continua sendo o mesmo.

Assim, na contagem

$4 \cdot 3 \cdot 2 = 24$

cada grupo de três pessoas foi contado **6 vezes**.

Para descobrir quantos grupos diferentes existem, precisamos retirar essa repetição:

$24 \div 6 = 4$

Portanto, é possível formar **4 grupos diferentes de três pessoas**.

> **De modo geral, quando escolhemos elementos e a ordem não importa, precisamos identificar quantas vezes cada grupo foi contado apenas por causa das diferentes ordens possíveis.**


```{admonition} Importante
:class: important

Se escolhemos $p$ elementos para formar um grupo, cada grupo pode aparecer em

$p!$
ordens diferentes.

Por isso, quando a ordem **não importa**, precisamos dividir a contagem feita por etapas sucessivas por \(p!\).

### Síntese

Entre Ana, Bruno, Caio e Davi, queremos escolher 3 pessoas para formar um grupo.

Se fizermos a contagem por etapas sucessivas:

- para a primeira escolha, temos 4 possibilidades;
- para a segunda escolha, restam 3 possibilidades;
- para a terceira escolha, restam 2 possibilidades.

Assim,

$$
4 \cdot 3 \cdot 2 = 24
$$

Mas cada grupo de 3 pessoas foi contado várias vezes, pois seus integrantes podem aparecer em diferentes ordens.

O número de ordens possíveis para 3 pessoas é

$$
3! = 3 \cdot 2 \cdot 1 = 6
$$

Portanto,

$$
\frac{4 \cdot 3 \cdot 2}{3!}
=
\frac{24}{6}
=
4
$$

Logo, podemos formar **4 grupos diferentes de três pessoas**.

```{admonition} Estratégia importante
:class: important

Quando diferentes ordens das mesmas escolhas representam o mesmo resultado:

1. **conte primeiro todos os caminhos possíveis**, como em um processo de escolhas sucessivas;

2. depois, identifique **quantas vezes cada resultado foi contado** apenas por causa das diferentes ordens possíveis;

3. **divida a contagem total por esse número de repetições**.

Assim, eliminamos as contagens repetidas e encontramos apenas os resultados realmente diferentes.

## Quando a ordem não altera o resultado

Nos problemas anteriores, escolhemos alguns elementos de um conjunto e observamos que diferentes ordens dos mesmos elementos podem representar o mesmo resultado.

Na linguagem tradicional da Análise Combinatória, a contagem das maneiras de escolher $p$ elementos entre $n$ elementos distintos, quando a ordem não altera o resultado e não há repetição, recebe o nome de **combinação simples**.

```{admonition} Para saber mais: combinação simples
:class: dropdown

Uma combinação simples é uma escolha de $p$ elementos entre $n$ elementos distintos, na qual diferentes ordens dos mesmos elementos representam o mesmo grupo.

Usando a notação $C_{n,p}$ para o número de combinações de $n$ elementos tomados $p$ a $p$,

$$
C_{n,p}=\frac{n!}{p!(n-p)!}.
$$

Essa fórmula apenas abrevia a estratégia utilizada anteriormente: contar as escolhas por etapas sucessivas e dividir pelo número de ordens possíveis dos elementos escolhidos.
```
