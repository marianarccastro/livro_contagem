# Quando cada escolha muda as possibilidades

Nos capítulos anteriores, vimos como contar processos formados por decisões sucessivas.

Neste capítulo, continuaremos usando essa ideia, mas agora cada escolha modifica as opções disponíveis na etapa seguinte, pois os mesmos elementos não podem ser usados novamente.

Como contar essas possibilidades de forma organizada?

## Organizando uma fila

Ana, Bruno e Caio precisam formar uma fila. De quantas maneiras diferentes essa fila pode ser organizada?

Para investigar essa situação, vamos nos colocar no papel de quem organiza a fila e pensar nas decisões que precisam ser tomadas.

A primeira decisão é escolher quem ocupará a primeira posição. Podemos escolher Ana, Bruno ou Caio.

Depois dessa escolha, restam duas pessoas para ocupar a segunda posição. Quando as duas primeiras posições estiverem preenchidas, a pessoa que restou ocupará a terceira posição.


## Representando as decisões

Podemos representar essas decisões em um diagrama de árvore. Cada ramificação indica uma escolha possível, considerando as escolhas que já foram feitas.

Acompanhe cada caminho desde o início até o final da árvore. Cada caminho completo representa uma maneira de organizar a fila.

```{mermaid}
flowchart LR
    I[Início]

    I --> A1[Ana]
    I --> B1[Bruno]
    I --> C1[Caio]

    A1 --> AB[Bruno]
    A1 --> AC[Caio]
    AB --> ABC[Caio]
    AC --> ACB[Bruno]

    B1 --> BA[Ana]
    B1 --> BC[Caio]
    BA --> BAC[Caio]
    BC --> BCA[Ana]

    C1 --> CA[Ana]
    C1 --> CB[Bruno]
    CA --> CAB[Bruno]
    CB --> CBA[Ana]

    ABC ~~~ FABC["Ana – Bruno – Caio"]
    ACB ~~~ FACB["Ana – Caio – Bruno"]

    BAC ~~~ FBAC["Bruno – Ana – Caio"]
    BCA ~~~ FBCA["Bruno – Caio – Ana"]

    CAB ~~~ FCAB["Caio – Ana – Bruno"]
    CBA ~~~ FCBA["Caio – Bruno – Ana"]

    classDef filaFinal fill:none,stroke:none,color:#000,font-weight:bold;
    class FABC,FACB,FBAC,FBCA,FCAB,FCBA filaFinal;
```


## Das decisões à contagem

O diagrama mostra que a organização da fila é formada por três decisões sucessivas:

- para a primeira posição, existem **3 possibilidades**;
- depois da primeira escolha, restam **2 possibilidades** para a segunda posição;
- depois das duas primeiras escolhas, resta **1 possibilidade** para a terceira posição.

Em vez de contar separadamente todos os caminhos da árvore, podemos multiplicar o número de possibilidades de cada etapa:

$$3\cdot2\cdot1=6.$$

Portanto, Ana, Bruno e Caio podem ser organizados em **6 filas diferentes**.

Neste problema, cada pessoa escolhida deixa de estar disponível para as posições seguintes. Por isso, o número de possibilidades diminui a cada etapa.

O diagrama de árvore permite visualizar essa redução. A multiplicação resume a quantidade de caminhos completos que podem ser formados.

Ao longo deste capítulo, utilizaremos essa mesma estratégia em situações nas quais cada escolha modifica as possibilidades seguintes, usando todos os elementos disponíveis ou apenas parte deles.

```{important}
Para resolver problemas desse tipo, coloque-se no papel de quem realiza a organização:

1. identifique quais decisões precisam ser tomadas;
2. determine quantas opções existem na primeira etapa;
3. a cada nova escolha, observe quais opções continuam disponíveis;
4. multiplique o número de possibilidades de cada etapa.
```

## Uma forma de abreviar o produto

Produtos formados por números naturais consecutivos e decrescentes aparecem com frequência em problemas de organização.

A expressão

$$5\cdot4\cdot3\cdot2\cdot1$$

pode ser escrita como

$$5!$$

e é lida como **cinco fatorial**.

```{admonition} Fatorial
:class: important

Para um número natural positivo $n$, definimos

$$n!=n\cdot(n-1)\cdot(n-2)\cdots2\cdot1.$$

Por exemplo,

$$4!=4\cdot3\cdot2\cdot1=24.$$

O fatorial é apenas uma forma abreviada de representar o produto decrescente. A contagem continua sendo explicada pelas escolhas sucessivas.
```

## Quando usamos todos os elementos

Nas filas e nos anagramas anteriores, utilizamos todos os elementos disponíveis. A cada escolha, um elemento deixa de estar disponível para as etapas seguintes, até que todos tenham sido utilizados.

Na linguagem tradicional da Análise Combinatória, a contagem das diferentes maneiras de organizar todos os elementos de um conjunto, sem repetição, recebe o nome de **permutação simples**.

```{admonition} Para saber mais: permutação simples
:class: dropdown

Uma permutação simples é uma organização de todos os $n$ elementos distintos disponíveis.

Para contar essas organizações, fazemos

$$n\cdot(n-1)\cdot(n-2)\cdots2\cdot1=n!.$$

Usando a notação $P_n$ para o número de permutações de $n$ elementos,

$$P_n=n!.$$

Essa fórmula apenas abrevia o produto obtido ao preencher todas as posições sucessivamente.
```

## Quando usamos apenas parte dos elementos

No problema do pódio, nem todos os atletas foram utilizados. Entre os 5 finalistas, fazemos apenas 3 escolhas.

Não é necessário continuar o produto até $1$, pois o processo termina quando todas as escolhas necessárias já foram feitas.

Compare:

Assim, a quantidade de fatores depende da quantidade de escolhas que precisam ser feitas.

## Quando usamos apenas parte dos elementos

Nos exemplos do pódio e da senha:

- havia $n$ elementos disponíveis;
- apenas $p$ posições foram preenchidas;
- cada elemento foi utilizado no máximo uma vez.

Nessas situações, utilizamos apenas parte dos elementos disponíveis. Na linguagem tradicional da Análise Combinatória, esse tipo de contagem recebe o nome de **arranjo simples**.

```{admonition} Para saber mais: arranjo simples
:class: dropdown

Para fazer $p$ escolhas sucessivas entre $n$ elementos distintos, sem repetição, o número de possibilidades é dado pelo produto de $p$ fatores:

$$n\cdot(n-1)\cdot(n-2)\cdots(n-p+1).$$

Esse produto pode ser abreviado por

$$A_{n,p}=\frac{n!}{(n-p)!}.$$

Por exemplo,

$$A_{5,3}=5\cdot4\cdot3=\frac{5!}{2!}=60.$$

Essa expressão não representa um novo princípio de contagem. Ela apenas sintetiza o produto obtido ao realizar sucessivamente $p$ escolhas, considerando que cada elemento escolhido deixa de estar disponível nas etapas seguintes.
```

## Até onde devemos multiplicar?

Compare as situações:

### Situação A

Cinco livros diferentes serão organizados em uma prateleira, utilizando todos os livros.

Como precisamos fazer 5 escolhas sucessivas,

$$5!=5\cdot4\cdot3\cdot2\cdot1=120.$$

### Situação B

Cinco livros diferentes estão disponíveis, mas apenas 3 serão escolhidos e colocados em três espaços numerados da prateleira.

Como precisamos fazer apenas 3 escolhas sucessivas,

$$5\cdot4\cdot3=60.$$

Nos dois casos, utilizamos o Princípio Fundamental da Contagem. A diferença está na quantidade de escolhas que precisam ser feitas até que o processo esteja completo.
