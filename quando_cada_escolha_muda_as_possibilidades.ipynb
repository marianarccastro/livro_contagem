```python
%%html
<div id="arvore-fila-cap3" class="arvore-fila-cap3">
  <div class="arvore-instrucoes">
    <strong>Construa a árvore:</strong> clique nos elementos disponíveis e complete todos os ramos de uma posição antes de passar à seguinte.
  </div>

  <div class="arvore-barra">
    <span id="arvore-etapa-cap3">Primeira posição</span>
    <button id="arvore-reiniciar-cap3" type="button">Reiniciar</button>
  </div>

  <div class="arvore-area" tabindex="0" aria-label="Diagrama de árvore">
    <svg id="arvore-svg-cap3" xmlns="http://www.w3.org/2000/svg"
         viewBox="0 0 675 430" role="img"
         aria-label="Árvore interativa para organizar Ana, Bruno e Caio em uma fila"></svg>
  </div>

  <div class="arvore-feedback" id="arvore-feedback-cap3" role="status" aria-live="polite"></div>

  <div class="arvore-reflexao" id="arvore-reflexao-cap3" hidden>
    <p><strong>Observe a árvore completa:</strong></p>
    <p>1. O que aconteceu com a quantidade de opções disponíveis a cada escolha?</p>
    <p>2. Quantas filas diferentes foram formadas?</p>
    <label for="arvore-resposta-cap3"><strong>Registre suas observações:</strong></label>
    <textarea id="arvore-resposta-cap3" placeholder="Escreva aqui..."></textarea>
  </div>
</div>

<style>
  .arvore-fila-cap3 {
    font-family: system-ui, -apple-system, BlinkMacSystemFont, "Segoe UI", sans-serif;
    color: #222;
    max-width: 900px;
    margin: 20px auto;
    padding: 18px;
    background: #fff;
    border: 1px solid #d6d9dc;
    border-radius: 12px;
    box-sizing: border-box;
  }
  .arvore-fila-cap3 .arvore-instrucoes {
    padding: 11px 13px;
    background: #eef5ff;
    border-left: 3px solid #4f7db8;
    border-radius: 6px;
    line-height: 1.5;
  }
  .arvore-fila-cap3 .arvore-barra {
    display: flex;
    align-items: center;
    justify-content: space-between;
    gap: 12px;
    margin: 15px 0 9px;
  }
  .arvore-fila-cap3 #arvore-etapa-cap3 { font-weight: 650; }
  .arvore-fila-cap3 button {
    padding: 8px 14px;
    color: #222;
    background: #fff;
    border: 1px solid #c6cbd1;
    border-radius: 7px;
    font: inherit;
    cursor: pointer;
  }
  .arvore-fila-cap3 button:hover { background: #f3f4f6; }
  .arvore-fila-cap3 button:focus-visible { outline: 3px solid #4f7db8; outline-offset: 2px; }
  .arvore-fila-cap3 .arvore-area {
    max-width: 100%;
    overflow-x: hidden;
    border: 1px solid #e5e7eb;
    border-radius: 8px;
    background: #fff;
  }
  .arvore-fila-cap3 svg {
    display: block;
    width: 100%;
    min-width: 0;
    height: auto;
  }
  .arvore-fila-cap3 svg .arvore-no.ativo { cursor: pointer; }
  .arvore-fila-cap3 svg .arvore-no.ativo:hover rect { stroke: #374151; stroke-width: 2.4; }
  .arvore-fila-cap3 svg .arvore-no.ativo:focus { outline: none; }
  .arvore-fila-cap3 svg .arvore-no.ativo:focus rect { stroke: #2563eb; stroke-width: 3; }
  .arvore-fila-cap3 svg .arvore-no.bloqueado { opacity: .62; }
  .arvore-fila-cap3 .arvore-feedback {
    padding: 12px 13px;
    min-height: 48px;
    line-height: 1.5;
    background: #f7f8fa;
    border: 1px solid #e5e7eb;
    border-radius: 8px;
    margin-top: 12px;
  }
  .arvore-fila-cap3 .arvore-reflexao { margin-top: 15px; line-height: 1.45; }
  .arvore-fila-cap3 .arvore-reflexao p { margin: 8px 0; }
  .arvore-fila-cap3 textarea {
    display: block;
    box-sizing: border-box;
    width: 100%;
    height: 100px;
    border: 1px solid #ccc;
    border-radius: 6px;
    font: inherit;
    padding: 10px;
    margin-top: 7px;
    resize: vertical;
  }
</style>

<script>
(() => {
  const container = document.getElementById("arvore-fila-cap3");
  if (!container) return;

  const svg = container.querySelector("#arvore-svg-cap3");
  const feedback = container.querySelector("#arvore-feedback-cap3");
  const etapa = container.querySelector("#arvore-etapa-cap3");
  const reflexao = container.querySelector("#arvore-reflexao-cap3");
  const resposta = container.querySelector("#arvore-resposta-cap3");
  const botaoReiniciar = container.querySelector("#arvore-reiniciar-cap3");
  const NS = "http://www.w3.org/2000/svg";

  const pessoas = ["Ana", "Bruno", "Caio"];
  const cores = {
    Ana: { fundo: "#fce7ed", borda: "#dda6b7" },
    Bruno: { fundo: "#e2efff", borda: "#94b6e2" },
    Caio: { fundo: "#e1f5e9", borda: "#9acbad" }
  };
  const caminhos = [
    { nomes: ["Ana", "Bruno", "Caio"], y: 78 },
    { nomes: ["Ana", "Caio", "Bruno"], y: 139 },
    { nomes: ["Bruno", "Ana", "Caio"], y: 200 },
    { nomes: ["Bruno", "Caio", "Ana"], y: 261 },
    { nomes: ["Caio", "Ana", "Bruno"], y: 322 },
    { nomes: ["Caio", "Bruno", "Ana"], y: 383 }
  ];
  const posPrimeira = { Ana: 108.5, Bruno: 230.5, Caio: 352.5 };
  const x = { inicio: 18, primeira: 118, segunda: 250, terceira: 382, resultado: 486 };
  const largura = { inicio: 68, nome: 72 };

  let fase = 0; // 0: início; 1: segunda posição; 2: terceira posição; 3: completo
  let primeirasAbertas = new Set();
  let segundasAbertas = new Set();

  function svgEl(tag, attrs = {}, texto) {
    const el = document.createElementNS(NS, tag);
    Object.entries(attrs).forEach(([k, v]) => el.setAttribute(k, v));
    if (texto !== undefined) el.textContent = texto;
    return el;
  }

  function linha(x1, y1, x2, y2) {
    svg.appendChild(svgEl("line", {
      x1, y1, x2, y2,
      stroke: "#9ca3af", "stroke-width": 1.7,
      "marker-end": "url(#seta-arvore-cap3)"
    }));
  }

  function texto(xPos, yPos, conteudo, attrs = {}) {
    svg.appendChild(svgEl("text", {
      x: xPos, y: yPos, fill: "#374151",
      "font-size": 15, "font-family": "system-ui, sans-serif",
      ...attrs
    }, conteudo));
  }

  function no(nome, xPos, yPos, clicavel, bloqueado, acao, descricao) {
    const cor = cores[nome] || { fundo: "#f3f4f6", borda: "#abb2bc" };
    const w = nome === "Início" ? largura.inicio : largura.nome;
    const grupo = svgEl("g", {
      class: "arvore-no" + (clicavel ? " ativo" : "") + (bloqueado ? " bloqueado" : "")
    });
    if (clicavel) {
      grupo.setAttribute("role", "button");
      grupo.setAttribute("tabindex", "0");
      grupo.setAttribute("aria-label", descricao);
      grupo.addEventListener("click", acao);
      grupo.addEventListener("keydown", (ev) => {
        if (ev.key === "Enter" || ev.key === " ") {
          ev.preventDefault();
          acao();
        }
      });
    }
    grupo.appendChild(svgEl("rect", {
      x: xPos, y: yPos - 18, width: w, height: 36, rx: 8,
      fill: cor.fundo, stroke: cor.borda, "stroke-width": 1.5
    }));
    grupo.appendChild(svgEl("text", {
      x: xPos + w / 2, y: yPos + 5, "text-anchor": "middle",
      fill: "#222", "font-size": 16, "font-family": "system-ui, sans-serif"
    }, nome));
    svg.appendChild(grupo);
  }

  function mensagem(conteudo) { feedback.textContent = conteudo; }
  function chave(c) { return c.nomes[0] + "-" + c.nomes[1]; }

  function abrirInicio() {
    fase = 1;
    mensagem("Para a primeira posição, há 3 opções: Ana, Bruno e Caio. Clique em cada nome para descobrir quais opções ficam disponíveis para a segunda posição.");
    desenhar();
  }

  function abrirPrimeira(nome) {
    if (fase !== 1 || primeirasAbertas.has(nome)) return;
    primeirasAbertas.add(nome);
    const restantes = pessoas.filter(p => p !== nome);
    if (primeirasAbertas.size === 3) {
      fase = 2;
      mensagem("Observe: depois de escolher quem ocupa a primeira posição, restam 2 opções para a segunda posição em cada caminho. Agora podemos avançar para a terceira posição.");
    } else {
      mensagem(`Se ${nome} ocupa a primeira posição, ${nome} deixa de estar disponível. Restam ${restantes[0]} e ${restantes[1]} para a segunda posição. Continue expandindo os outros ramos.`);
    }
    desenhar();
  }

  function abrirSegunda(caminho) {
    if (fase !== 2 || segundasAbertas.has(chave(caminho))) return;
    segundasAbertas.add(chave(caminho));
    const [a, b, c] = caminho.nomes;
    if (segundasAbertas.size === caminhos.length) {
      fase = 3;
      mensagem("Você completou a árvore! Para a primeira posição havia 3 opções; depois de cada primeira escolha, restavam 2; para a última posição, restava apenas 1. Cada caminho completo representa uma fila diferente.");
    } else {
      mensagem(`No caminho ${a} → ${b}, essas duas pessoas já foram escolhidas. Resta apenas 1 opção para a terceira posição: ${c}.`);
    }
    desenhar();
  }

  function desenhar() {
    svg.replaceChildren();
    const defs = svgEl("defs");
    const marker = svgEl("marker", {
      id: "seta-arvore-cap3", viewBox: "0 0 10 10", refX: 9,
      refY: 5, markerWidth: 5, markerHeight: 5, orient: "auto-start-reverse"
    });
    marker.appendChild(svgEl("path", { d: "M 0 0 L 10 5 L 0 10 z", fill: "#9ca3af" }));
    defs.appendChild(marker);
    svg.appendChild(defs);

    texto(154, 34, "1ª posição", { "text-anchor": "middle", "font-weight": 600 });
    texto(286, 34, "2ª posição", { "text-anchor": "middle", "font-weight": 600 });
    texto(418, 34, "3ª posição", { "text-anchor": "middle", "font-weight": 600 });
    texto(568, 34, "Fila formada", { "text-anchor": "middle", "font-weight": 600 });

    if (fase >= 1) {
      pessoas.forEach(p => linha(x.inicio + largura.inicio, 230.5, x.primeira - 6, posPrimeira[p]));
    }
    caminhos.forEach(c => {
      if (primeirasAbertas.has(c.nomes[0])) {
        linha(x.primeira + largura.nome, posPrimeira[c.nomes[0]], x.segunda - 6, c.y);
      }
      if (segundasAbertas.has(chave(c))) {
        linha(x.segunda + largura.nome, c.y, x.terceira - 6, c.y);
      }
    });

    no("Início", x.inicio, 230.5, fase === 0, false, abrirInicio,
      "Mostrar opções para a primeira posição");

    if (fase >= 1) {
      pessoas.forEach(p => no(p, x.primeira, posPrimeira[p],
        fase === 1 && !primeirasAbertas.has(p), false,
        () => abrirPrimeira(p), `Mostrar opções para a segunda posição após escolher ${p}`));
    }

    caminhos.forEach(c => {
      if (primeirasAbertas.has(c.nomes[0])) {
        const ativado = segundasAbertas.has(chave(c));
        no(c.nomes[1], x.segunda, c.y, fase === 2 && !ativado,
          fase === 1, () => abrirSegunda(c),
          `Mostrar a terceira posição depois de ${c.nomes[0]} e ${c.nomes[1]}`);
      }
      if (segundasAbertas.has(chave(c))) {
        no(c.nomes[2], x.terceira, c.y, false, false, null, "");
        texto(x.resultado, c.y + 5, c.nomes.join(" – "), { "font-size": 15 });
      }
    });

    etapa.textContent = ["Primeira posição", "Segunda posição", "Terceira posição", "Árvore completa"][fase];
    reflexao.hidden = fase !== 3;
  }

  botaoReiniciar.addEventListener("click", () => {
    fase = 0;
    primeirasAbertas = new Set();
    segundasAbertas = new Set();
    resposta.value = "";
    mensagem("Clique em Início para descobrir quem pode ocupar a primeira posição da fila.");
    desenhar();
  });
  mensagem("Clique em Início para descobrir quem pode ocupar a primeira posição da fila.");
  desenhar();
})();
</script>
```

## Das decisões à contagem

Ao construir o diagrama de árvore, observamos que a organização da fila envolve três decisões sucessivas:

- **Primeira posição:** 3 opções disponíveis.
- **Segunda posição:** 2 opções disponíveis, pois uma pessoa já foi escolhida.
- **Terceira posição:** 1 opção disponível, pois duas pessoas já foram escolhidas.

Em vez de contar os caminhos completos da árvore um a um, podemos utilizar o **Princípio Fundamental da Contagem**, multiplicando a quantidade de opções disponíveis em cada etapa:

$$3\cdot2\cdot1=6.$$

Portanto, Ana, Bruno e Caio podem ser organizados em **6 filas diferentes**.

:::{tip} Estratégia
Para resolver problemas desse tipo, coloque-se no papel de quem realiza a organização e toma as decisões:

1. Identifique quais decisões precisam ser tomadas.
2. Determine quantas opções estão disponíveis na primeira etapa.
3. A cada nova escolha, observe quais opções continuam disponíveis.
4. Utilize o Princípio Fundamental da Contagem, multiplicando a quantidade de opções disponíveis em cada etapa.
:::

### Uma forma de abreviar o produto

Para representar esses produtos de maneira mais simples, utilizamos a notação de fatorial.

:::{important} 🔷 Definição
:icon: false

**Fatorial**

Para um número natural positivo $n$, definimos

$$n!=n\cdot(n-1)\cdot(n-2)\cdots 2\cdot1.$$

Por exemplo,

$$4!=4\cdot3\cdot2\cdot1=24.$$

Lemos $4!$ como **quatro fatorial**.
:::

Considere a palavra **AMOR**. Quantos anagramas diferentes podem ser formados com suas letras?

### Um nome para essas organizações

Nas filas e nos anagramas, organizamos todos os elementos disponíveis, sem repetir nenhum deles. Na Análise Combinatória, cada uma dessas organizações recebe o nome de **permutação simples**.

:::{seealso} Para saber mais: permutação simples

Uma **permutação simples** é uma ordenação de todos os elementos de um conjunto de elementos distintos.

O número de permutações simples de $n$ elementos distintos é representado por $P_n$. Como já vimos, essa quantidade pode ser calculada pelo fatorial:

$$P_n=n!$$

:::

## E quando não precisamos usar todos os elementos?

### Formando um pódio

Em uma corrida, cinco atletas disputam as três primeiras posições do pódio: ouro, prata e bronze.

De quantas maneiras diferentes o pódio pode ser formado?

Para organizar a contagem, pense nas escolhas necessárias:

- Quantos atletas podem receber a medalha de ouro? Depois dessa escolha, quantas opções restam para a prata? E para o bronze?

```python
%%html
<p><b>Registre a contagem:</b></p>
<textarea style="width: 100%; height: 100px; border: 1px solid #ccc; border-radius: 4px;"></textarea>
```

:::{tip} Solução
:class: dropdown

Há 5 opções para a medalha de ouro. Depois dessa escolha, restam 4 atletas para a medalha de prata e 3 para a medalha de bronze.

Logo,

$$5\cdot4\cdot3=60.$$

Podem ser formados 60 pódios diferentes.

Observe que o processo termina quando as três medalhas já foram distribuídas.
:::


### Comparando as organizações

Considere duas situações envolvendo cinco pessoas:

- **Organizar todas em uma fila:** utilizamos os cinco elementos disponíveis. Como vimos, o número de permutações simples pode ser calculado pelo fatorial:

  $$P_5=5!=5\cdot4\cdot3\cdot2\cdot1=120.$$

- **Definir os três primeiros colocados de uma competição:** temos cinco atletas disponíveis, mas apenas três serão escolhidos para receber as medalhas de ouro, prata e bronze. Nesse caso, utilizamos o Princípio Fundamental da Contagem:

  $$5\cdot4\cdot3=60.$$

**Na permutação simples, todos os elementos disponíveis são organizados.** Quando utilizamos apenas parte deles, como no pódio, a contagem termina assim que todas as escolhas necessárias foram realizadas.

Em ambas as situações, organizamos a contagem por meio de decisões sucessivas.
